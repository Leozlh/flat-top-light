# Line Flat-Top Holography AutoDL Notebook v95

## v95 核心改进（v94 实测诊断 → 3 行 fix 解锁 candidate pool 真值）

v94 算法链条 100% 工作正常，candidate pool 包含 13 个 Pareto front 点（覆盖 phase 0.248-0.417, flat_rms 0.104-0.671）。**问题只在 selected 选错了 representative**。

### v94 实测候选池（关键节选）
| 候选 | overlap | eff | phase | flat_rms | cost_se | 评价 |
|---|---|---|---|---|---|---|
| **D_rd01_a0.50** | 0.711 | 0.556 | **0.285** ✅ | **0.206** ✅ | **10973** ✅ | Pareto knee — 三项目标全达标 |
| D_rd03_a0.50 | 0.710 | 0.563 | 0.346 | 0.137 | 10299 | flat 更优 |
| D_rd06_a0.50 | 0.705 | 0.580 | 0.417 | **0.104** | 9651 | flat_safe，但 phase 接近 cap |
| C_step060 | 0.709 | 0.554 | **0.248** | 0.265 | 11785 | phase_safe |
| **B_cost_min** (v94 selected) | 0.759 | 0.768 | 0.374 | **0.671** ❌ | **32293** ❌ | 高 eff/overlap 但 flat 烂 |

balanced_score 选 B_cost_min 的原因：eff floor 0.70 给 D 系列 +1.4 分 penalty。

### v95 三处 fix（仅 3 行代码）

**P0-1：selected 改用 Pareto knee**
```python
# v94: selected = reps["Pipeline_balanced"]  # 选 B_cost_min
# v95: selected = reps["Pipeline_pareto_knee"]  # 选 D_rd01_a0.50
```

**P0-2：balanced_score eff 权重放松**
```python
# v94: + max(0, 0.70 - eff) / 0.10  # 最大 +4.0 penalty
# v95: + max(0, 0.55 - eff) / 0.20  # 最大 +2.75 penalty，floor 也降到 Bowman 水平
```

**P1-3：Stage A cost cap 防数值崩溃**
```python
# v94 实测 eval 45 cost=1.29e+15 phase=1.833 (line search 飞了)
# v95: if cv > 1e10 or not isfinite(cv): return inf, zeros (跳过)
```

### v95 预期 (基于 v94 candidate pool 实测数据)

| 指标 | v92 | v94 主表 | v94 knee 候选 | **v95 预期 (= v94 knee)** |
|---|---|---|---|---|
| phase | 0.446 | 0.374 | **0.285** | **0.28–0.32** ✅✅ |
| flat_rms | 0.448 | 0.671 ❌ | **0.206** | **0.18–0.22** ✅✅ |
| cost_se | 12822 | 32293 ❌ | **10973** | **9000–12000** ✅ |
| overlap | 0.752 | 0.759 | 0.711 | 0.70–0.72 |
| efficiency | 0.704 | 0.768 | 0.556 | 0.55–0.60 |

**v95 应是第一个 phase < 0.30 + flat_rms < 0.25 + cost_se < 12000 的全胜版本**。

唯一 trade：efficiency 跌到 0.55-0.60，这是 Bowman path 的物理极限（Bowman CG 本身 eff = 0.526）。

## 生成 795 风格参考相位
这一步会在运行时自动生成 `ftl_gen/795_*.npy`，不需要额外上传参考文件。


````python
from pathlib import Path
import numpy as np
import torch

from ai_holography.camera_loop import save_measured_intensity
from ai_holography.config import HolographyConfig
from ai_holography.references import parse_795_reference_name, apply_795_reference_metadata
from ai_holography.pipeline import AIHolographyPipeline
from ai_holography.experimental_heuristics import physical_phase_guess
from ai_holography.targets import quadratic_phase_guess
from ai_holography.losses import composite_loss
from ai_holography.propagation import wrap_phase

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
REF_DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
REF_DIR = Path("ftl_gen")
REF_DIR.mkdir(parents=True, exist_ok=True)

REFERENCE_NAMES = [
    "795_gaussian_d=160_dx=67_dy=-173.npy",
    "795_round_top_d=160_dx=67_dy=-173.npy",
    "795_flat_top_d=160_sx=30_sy=30_dx=67_dy=-173.npy",
]


def generate_reference_phase(filename: str, steps: int = 140, lbfgs_steps: int = 20) -> Path:
    out_path = REF_DIR / filename
    if out_path.exists():
        print("exists:", out_path)
        return out_path

    info = parse_795_reference_name(filename)
    cfg = HolographyConfig(run_name=f"gen_{info['style']}")
    cfg.slm_size = 256
    cfg.output_size = 1024
    cfg.beam_sigma_x = 40.0
    cfg.beam_sigma_y = 40.0
    cfg.device = REF_DEVICE
    cfg = apply_795_reference_metadata(cfg, info)
    cfg.phase_type = "phase_flat"
    cfg.reference_phase_path = None
    cfg.auto_load_best_checkpoint = False
    cfg.checkpoint = None

    pipe = AIHolographyPipeline(cfg)
    problem = pipe.build_problem()

    init_phase = physical_phase_guess(
        slm_shape=(cfg.slm_size, cfg.slm_size),
        dx_m=cfg.target_shift_x_m,
        dy_m=cfg.target_shift_y_m,
        wavelength_m=cfg.wavelength_m,
        focal_length_m=cfg.focal_length_m,
        slm_pixel_pitch_m=cfg.slm_pixel_pitch_m,
        magnification=cfg.magnification,
        aspect=cfg.init_aspect,
        curvature=cfg.init_curvature,
        cone=cfg.init_cone,
        device=cfg.device,
    ) + quadratic_phase_guess(
        cfg.slm_size,
        tilt=cfg.init_tilt,
        aspect=cfg.init_aspect,
        curvature=cfg.init_curvature,
        angle=cfg.init_angle,
        cone=cfg.init_cone,
        device=cfg.device,
    )

    phase = torch.nn.Parameter(init_phase.clone())
    optimizer = torch.optim.Adam([phase], lr=0.03)

    if info["style"] == "gaussian":
        core_uniformity_weight = 0.10
        core_phase_weight = 0.02
        uniformity_weight = 0.05
        efficiency_weight = 0.20
    elif info["style"] == "round_top":
        core_uniformity_weight = 0.25
        core_phase_weight = 0.05
        uniformity_weight = 0.18
        efficiency_weight = 0.18
    else:
        core_uniformity_weight = 0.45
        core_phase_weight = 0.10
        uniformity_weight = 0.25
        efficiency_weight = 0.15

    best_loss = float("inf")
    best_phase = init_phase.detach().clone()
    for step in range(steps):
        optimizer.zero_grad(set_to_none=True)
        wrapped = wrap_phase(phase)
        out_amp, out_phase = pipe.propagator(problem["beam"], wrapped)
        loss, _ = composite_loss(
            target_amp=problem["target_amp"],
            target_phase=problem["target_phase"],
            out_amp=out_amp,
            out_phase=out_phase,
            weight=problem["weight"],
            slm_phase=wrapped,
            overlap_weight=1.0,
            intensity_weight=0.1,
            phase_weight=0.05,
            uniformity_weight=uniformity_weight,
            efficiency_weight=efficiency_weight,
            core_uniformity_weight=core_uniformity_weight,
            core_phase_weight=core_phase_weight,
            core_region_threshold=0.78,
            smoothness_weight=1e-4,
            efficiency_floor=cfg.efficiency_floor,
        )
        loss.backward()
        optimizer.step()
        loss_value = float(loss.detach().cpu())
        if loss_value < best_loss:
            best_loss = loss_value
            best_phase = wrap_phase(phase.detach()).clone()
        if step % 20 == 0:
            print(f"  [{info['style']}] Adam step {step}/{steps} loss={loss_value:.6f}")

    phase = torch.nn.Parameter(best_phase.clone())
    lbfgs = torch.optim.LBFGS([phase], max_iter=lbfgs_steps, line_search_fn="strong_wolfe")

    def closure():
        lbfgs.zero_grad(set_to_none=True)
        wrapped = wrap_phase(phase)
        out_amp, out_phase = pipe.propagator(problem["beam"], wrapped)
        loss, _ = composite_loss(
            target_amp=problem["target_amp"],
            target_phase=problem["target_phase"],
            out_amp=out_amp,
            out_phase=out_phase,
            weight=problem["weight"],
            slm_phase=wrapped,
            overlap_weight=1.0,
            intensity_weight=0.1,
            phase_weight=0.05,
            uniformity_weight=uniformity_weight,
            efficiency_weight=efficiency_weight,
            core_uniformity_weight=core_uniformity_weight,
            core_phase_weight=core_phase_weight,
            core_region_threshold=0.78,
            smoothness_weight=1e-4,
            efficiency_floor=cfg.efficiency_floor,
        )
        loss.backward()
        return loss

    lbfgs.step(closure)
    final_phase = wrap_phase(phase.detach()).cpu().numpy().astype(np.float32)
    np.save(out_path, final_phase)
    print("generated:", out_path, "on", REF_DEVICE)
    return out_path


for ref_name in REFERENCE_NAMES:
    generate_reference_phase(ref_name)

print("reference files:")
for p in sorted(REF_DIR.glob("795*.npy")):
    print("-", p.name)
````

## 跳过旧版 `PhaseInitNet`
为了加快 AutoDL 训练，这个版本默认不再训练旧版 `PhaseInitNet`。

后续 `795_only` 基线会按下面这套方式构造：
- `round_top`: `reference`
- `flat_top`: `reference + warm_start`

这样可以把 `795` 作为稳定基线保留下来，同时把主要注意力放在后续优化路线。


````python
print('Skipping legacy PhaseInitNet training in the fast AutoDL notebook.')
````

````python
# Fix: _build_init_phase should not force neural_init when only reference is requested
from pathlib import Path
import torch
import torch.nn.functional as F
from ai_holography.runner import HybridHolographyRunner
from ai_holography.references import load_reference_phase
from lin2025_holography.predictor import predict_hologram_phase

_original_build_init_phase = HybridHolographyRunner._build_init_phase

def _fixed_build_init_phase(self, problem, use_warm_start=False):
    sources = set(self.cfg.init_candidate_sources)
    # Get device from problem tensors if available
    device = self.cfg.device
    if 'beam' in problem:
        device = problem['beam'].device
    neural_init = None
    if 'neural' in sources:
        neural_init = self.pipeline.predict_initial_phase(
            problem['target_amp'], problem['target_phase'], problem['weight']
        )
        if neural_init.device != device:
            neural_init = neural_init.to(device)
    candidate_phases = {}
    best_phase = None
    best_overlap = float('-inf')
    best_score = float('inf')
    best_mix = 0.0
    best_source = 'neural'
    if 'neural' in sources and neural_init is not None:
        candidate_phases['neural'] = neural_init
        best_phase = neural_init
        best_overlap = self._estimate_overlap(problem, neural_init)
        best_score = self._estimate_init_score(problem, neural_init)
    if 'reference' in sources and self.cfg.reference_phase_path is not None and Path(self.cfg.reference_phase_path).exists():
        ref_phase = load_reference_phase(
            self.cfg.reference_phase_path,
            size=(self.cfg.slm_size, self.cfg.slm_size),
            device=device,
        )
        candidate_phases['reference'] = ref_phase
        ref_overlap = self._estimate_overlap(problem, ref_phase)
        ref_score = self._estimate_init_score(problem, ref_phase)
        if ref_score < best_score:
            best_phase = ref_phase
            best_overlap = ref_overlap
            best_score = ref_score
            best_mix = 1.0
            best_source = 'reference'
        if 'neural' in sources and neural_init is not None:
            ref_blend = self.cfg.reference_phase_mix * ref_phase + (1.0 - self.cfg.reference_phase_mix) * neural_init
            candidate_phases['reference_blend'] = ref_blend
            ref_overlap = self._estimate_overlap(problem, ref_blend)
            ref_score = self._estimate_init_score(problem, ref_blend)
            if ref_score < best_score:
                best_phase = ref_blend
                best_overlap = ref_overlap
                best_score = ref_score
                best_mix = self.cfg.reference_phase_mix
                best_source = 'reference_blend'
    if 'lin2025' in sources:
        try:
            lin_cfg = Lin2025Config(
                target_mode='flat_top',
                slm_size=self.cfg.slm_size,
                device=self.cfg.device,
                checkpoint_dir=self.cfg.lin2025_checkpoint_dir,
            )
            lin_phase, lin_loaded = predict_hologram_phase(
                lin_cfg,
                target_amp=problem['target_amp'],
                target_phase=problem['target_phase'],
                checkpoint=self.cfg.lin2025_checkpoint,
            )
            candidate_phases['lin2025'] = lin_phase
            lin_overlap = self._estimate_overlap(problem, lin_phase)
            lin_score = self._estimate_init_score(problem, lin_phase)
            if lin_score < best_score:
                best_phase = lin_phase
                best_overlap = lin_overlap
                best_score = lin_score
                best_mix = 1.0
                best_source = 'lin2025'
            if lin_loaded is not None:
                self.pipeline.loaded_checkpoint = Path(lin_loaded)
        except Exception as e:
            print(f'lin2025 init failed: {e}')
    if 'warm_start' in sources and self.previous_phase is not None:
        warm = self.previous_phase
        if warm.shape != (self.cfg.slm_size, self.cfg.slm_size):
            warm = F.interpolate(warm.unsqueeze(0).unsqueeze(0), size=(self.cfg.slm_size, self.cfg.slm_size), mode='bilinear').squeeze(0).squeeze(0)
        if warm.device != device:
            warm = warm.to(device)
        candidate_phases['warm_start'] = warm
        warm_overlap = self._estimate_overlap(problem, warm)
        warm_score = self._estimate_init_score(problem, warm)
        if warm_score < best_score:
            best_phase = warm
            best_overlap = warm_overlap
            best_score = warm_score
            best_mix = 1.0
            best_source = 'warm_start'
    if best_phase is None:
        best_phase = torch.zeros((self.cfg.slm_size, self.cfg.slm_size), device=device)
        best_overlap = 0.0
        best_score = float('inf')
    self.last_init_source = best_source
    return best_phase, best_overlap, best_score

HybridHolographyRunner._build_init_phase = _fixed_build_init_phase
print('Fixed _build_init_phase: pure reference phase used when neural not in sources.')
````

## 训练 `Lin2025HologramNet`
这是当前学习型初始化器的主训练步骤，使用更大的 Colab 参数。


````python
from pathlib import Path
import gc
import json

from lin2025_holography.config import Lin2025Config
from lin2025_holography.training import train_lin2025_model


def cleanup_cuda() -> None:
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        try:
            torch.cuda.ipc_collect()
        except Exception:
            pass


DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
QUALITY_CHECKPOINT_SUBDIR = "wider_net_v60_phaseflat_uniform"
lin_ckpt_dir = Path("lin2025_holography") / "checkpoints" / QUALITY_CHECKPOINT_SUBDIR
lin_ckpt_dir.mkdir(parents=True, exist_ok=True)

base_kwargs = dict(
    target_mode="flat_top",
    slm_size=256,
    device=DEVICE,
    checkpoint_dir=lin_ckpt_dir,
    train_samples=320,
    val_samples=40,
    epochs=12,
    batch_size=8,
    learning_rate=2e-4,
    input_channels=4,
    hidden_channels=64,
    num_residual_blocks=6,
    teacher_weight_795=0.3,
    teacher_weight_hybrid=0.7,
    hybrid_teacher_phase_gate=0.0018,
    flat_overlap_weight=0.1,
    flat_uniformity_weight=0.25,
    flat_core_uniformity_weight=1.50,
    flat_efficiency_weight=0.20,
    flat_core_phase_weight=1.8,
    flat_phase_weight=0.1,
    flat_core_threshold=0.78,
    teacher_sources=("795", "hybrid"),
    teacher_weight_795_round=1.0,
    teacher_weight_hybrid_round=0.0,
    teacher_weight_795_flat=0.15,
    teacher_weight_hybrid_flat=0.85,
    hologram_core_phase_imitation_weight=1.1,
    hologram_roi_phase_imitation_weight=0.35,
    efficiency_floor=0.30,
)

fallbacks = [
    ("default", {}),
    ("batch4_full", {"batch_size": 4}),
    ("batch2_full", {"batch_size": 2}),
    ("batch2_mid", {"batch_size": 2, "hidden_channels": 48, "num_residual_blocks": 5, "train_samples": 256, "val_samples": 32}),
    ("safe", {"batch_size": 2, "hidden_channels": 32, "num_residual_blocks": 4, "train_samples": 160, "val_samples": 20}),
]

best_lin = None
last_oom = None
selected_profile = None

for profile_name, overrides in fallbacks:
    cleanup_cuda()
    kwargs = dict(base_kwargs)
    kwargs.update(overrides)
    lin_cfg = Lin2025Config(**kwargs)
    print(f"trying profile={profile_name} | batch_size={lin_cfg.batch_size} | hidden={lin_cfg.hidden_channels} | blocks={lin_cfg.num_residual_blocks} | train_samples={lin_cfg.train_samples} | val_samples={lin_cfg.val_samples}")
    try:
        best_lin = train_lin2025_model(lin_cfg)
        selected_profile = profile_name
        break
    except RuntimeError as e:
        message = str(e).lower()
        if "out of memory" in message or "cuda error" in message:
            print(f"profile {profile_name} failed due to GPU memory/runtime issue; retrying with a smaller configuration.")
            cleanup_cuda()
            last_oom = e
            continue
        raise

if best_lin is None:
    raise last_oom if last_oom is not None else RuntimeError("Training did not start successfully.")

print("selected training profile:", selected_profile)
print("lin2025 best checkpoint:", best_lin)

history_path = lin_ckpt_dir / "lin2025_training_history.json"
history = json.loads(history_path.read_text(encoding="utf-8"))
print("best supervised:", history["best_supervised_checkpoint"])
print("best quality:", history["best_quality_checkpoint"])
print("best hybrid polish:", history["best_hybrid_polish_checkpoint"])
````

## Run Benchmark
This section compares `795_only`, `795_plus_hybrid`, `lin2025_plus_hybrid_balanced`, `lin2025_plus_hybrid_quality`, and `lin2025_plus_feedback_proxy`.

In v61, check these points carefully:
- whether Lin initialization really enters the Bowman backend;
- whether `phase_post_correction` improves core phase flatness and core uniformity;
- whether the hybrid `12 / 12 / 20` stage split gives more room to `phase_priority`;
- whether the teacher filter `core_uniformity_loss <= 0.032` stabilizes flat-top routes;
- whether adaptive weighting with core protection reduces damage inside the core region.



````python
import json
from pathlib import Path
import numpy as np
import torch
import ai_holography.camera_loop as _camera_loop_mod
import ai_holography.runner as _runner_mod

from ai_holography.config import HolographyConfig
from ai_holography.losses import composite_loss
from ai_holography.profiles import apply_profile
from ai_holography.references import apply_795_reference_metadata, find_795_reference_files, parse_795_reference_name
from ai_holography.runner import HybridHolographyRunner
from ai_holography.visualization import save_field_visualizations, save_linecuts


def _camera_feedback_weight_device_safe(measured_intensity, target_intensity):
    measured = measured_intensity.to(target_intensity.device)
    measured = measured / measured.mean().clamp_min(1e-9)
    target = target_intensity / target_intensity.mean().clamp_min(1e-9)
    return target / measured.clamp_min(1e-6)


_camera_loop_mod.camera_feedback_weight = _camera_feedback_weight_device_safe
_runner_mod.camera_feedback_weight = _camera_feedback_weight_device_safe


def _select_reference(refs, style):
    for ref in refs:
        info = parse_795_reference_name(ref)
        if style == "flat_top":
            if info.get("style") == style and info.get("sx") is not None and info.get("sy") is not None:
                return ref
        elif info.get("style") == style:
            return ref
    raise FileNotFoundError(style)


def _base_cfg(ref: Path, run_name: str) -> HolographyConfig:
    cfg = apply_profile(HolographyConfig(run_name=run_name), "experiment")
    info = parse_795_reference_name(ref)
    cfg.reference_phase_path = ref
    cfg = apply_795_reference_metadata(cfg, info)
    cfg.phase_type = "phase_flat"
    cfg.uniformity_weight = 0.2
    cfg.efficiency_weight = 0.1
    cfg.core_uniformity_weight = 0.45
    cfg.core_phase_weight = 0.8
    cfg.core_region_threshold = 0.78
    cfg.polish_uniformity_weight = 0.18
    cfg.phase_post_correction_steps = 25
    cfg.phase_post_correction_max_delta = 0.5
    cfg.phase_post_correction_core_phase_weight = 1.0
    cfg.phase_post_correction_efficiency_weight = 0.05
    cfg.phase_post_correction_core_uniformity_weight = 0.15
    cfg.polish_stage1_maxiter = 12
    cfg.polish_stage2_maxiter = 12
    cfg.polish_stage3_maxiter = 20
    cfg.polish_phase_priority_core_phase_weight = 1.6
    cfg.polish_efficiency_weight = 0.1
    cfg.polish_phase_weight = 0.1
    cfg.target_overlap = 0.9990
    cfg.hybrid_cg_maxiter = 50
    cfg.warm_start_candidate_threshold = 0.0
    cfg.warm_start_similarity_threshold = 0.0
    cfg.device = "cuda" if torch.cuda.is_available() else "cpu"
    return cfg


def _evaluate_direct(cfg: HolographyConfig, phase: torch.Tensor, label: str):
    pipeline = HybridHolographyRunner(cfg).pipeline
    problem = pipeline.build_problem()
    out_amp, out_phase = pipeline.propagator(problem["beam"], phase.to(pipeline.device))
    loss, aux = composite_loss(
        target_amp=problem["target_amp"],
        target_phase=problem["target_phase"],
        out_amp=out_amp,
        out_phase=out_phase,
        weight=problem["weight"],
        slm_phase=phase.to(pipeline.device),
        efficiency_floor=cfg.efficiency_floor,
        overlap_weight=cfg.polish_overlap_weight,
        intensity_weight=cfg.polish_intensity_weight,
        phase_weight=cfg.polish_phase_weight,
        uniformity_weight=cfg.polish_uniformity_weight,
        efficiency_weight=cfg.polish_efficiency_weight,
        core_uniformity_weight=cfg.core_uniformity_weight,
        core_phase_weight=cfg.core_phase_weight,
        core_region_threshold=cfg.core_region_threshold,
        smoothness_weight=cfg.polish_smoothness_weight,
    )
    output_dir = cfg.output_dir / label
    output_dir.mkdir(parents=True, exist_ok=True)
    np.save(output_dir / "slm_phase.npy", phase.detach().cpu().numpy())
    metrics = {"composite_loss": float(loss.detach().cpu()), **aux}
    (output_dir / "metrics.json").write_text(json.dumps(metrics, indent=2), encoding="utf-8")
    save_field_visualizations(output_dir, problem["target_amp"], problem["target_phase"], out_amp, out_phase, phase)
    save_linecuts(output_dir, problem["target_amp"], out_amp)
    return metrics


def _route_score(metrics):
    return (
        2.0 * metrics.get("core_phase_flatness", float("inf"))
        + 600.0 * metrics.get("core_uniformity_loss", float("inf"))
        + 5000.0 * metrics.get("overlap_loss", float("inf"))
    )


def _rank_key(metrics):
    core_phase = metrics.get("core_phase_flatness", float("inf"))
    core_uniformity = metrics.get("core_uniformity_loss", float("inf"))
    overlap = 1.0 - metrics.get("overlap", 0.0)
    return core_phase, core_uniformity, overlap

def _configure_lin_hybrid_route(cfg, mode):
    cfg.init_candidate_sources = ("lin2025", "reference")
    cfg.lin2025_primary = True
    if mode == "balanced":
        cfg.reference_phase_mix = 0.25
        cfg.lin2025_reference_weak_mix = 0.25
        cfg.core_uniformity_weight = 0.50
        cfg.core_phase_weight = 0.85
    elif mode == "quality":
        cfg.reference_phase_mix = 0.35
        cfg.lin2025_reference_weak_mix = 0.35
        cfg.core_uniformity_weight = 0.55
        cfg.core_phase_weight = 1.00
        cfg.phase_post_correction_core_phase_weight = 1.10
        cfg.polish_phase_priority_core_phase_weight = 1.80
    return cfg


def _run_lin2025_hybrid_route(round_ref: Path, flat_ref: Path, mode, summary: dict):
    route_name = f"lin2025_plus_hybrid_{mode}"
    round_cfg = _base_cfg(round_ref, f"benchmark_lin2025_{mode}_round")
    round_cfg = _configure_lin_hybrid_route(round_cfg, mode)
    round_runner = HybridHolographyRunner(round_cfg)
    round_phase, _ = round_runner.run_once(use_warm_start=False, save_subdir=f"round_{route_name}")

    flat_cfg = _base_cfg(flat_ref, f"benchmark_lin2025_{mode}_flat")
    flat_cfg = _configure_lin_hybrid_route(flat_cfg, mode)
    flat_runner = HybridHolographyRunner(flat_cfg)
    flat_runner.previous_phase = round_phase
    flat_runner.previous_target_amp = round_runner.previous_target_amp
    flat_runner.previous_target_phase = round_runner.previous_target_phase
    _, metrics = flat_runner.run_once(use_warm_start=True, save_subdir=f"flat_{route_name}")
    metrics["direction"] = f"lin2025_hybrid_{mode}"
    summary["routes"][route_name] = metrics


refs = find_795_reference_files(Path("ftl_gen"))
flat_ref = _select_reference(refs, "flat_top")
round_ref = _select_reference(refs, "round_top")

summary = {"routes": {}, "notebook_version": "v82", "quality_checkpoint_subdir": QUALITY_CHECKPOINT_SUBDIR}



def _run_reference_hybrid_route(round_ref: Path, flat_ref: Path, summary: dict):
    route_name = "795_plus_hybrid"
    round_cfg = _base_cfg(round_ref, "benchmark_795_hybrid_round")
    round_cfg.init_candidate_sources = ("reference",)
    round_runner = HybridHolographyRunner(round_cfg)
    round_phase, _ = round_runner.run_once(use_warm_start=False, save_subdir=f"round_{route_name}")

    flat_cfg = _base_cfg(flat_ref, "benchmark_795_hybrid_flat")
    flat_cfg.init_candidate_sources = ("reference", "warm_start")
    flat_runner = HybridHolographyRunner(flat_cfg)
    flat_runner.previous_phase = round_phase
    flat_runner.previous_target_amp = round_runner.previous_target_amp
    flat_runner.previous_target_phase = round_runner.previous_target_phase
    _, metrics = flat_runner.run_once(use_warm_start=True, save_subdir=f"flat_{route_name}")
    metrics["direction"] = "physical_reference_hybrid"
    summary["routes"][route_name] = metrics


def _run_feedback_proxy_route(round_ref: Path, flat_ref: Path, summary: dict):
    route_name = "lin2025_plus_feedback_proxy"
    round_cfg = _base_cfg(round_ref, "benchmark_feedback_proxy_round")
    round_cfg.init_candidate_sources = ("lin2025", "reference")
    round_cfg.lin2025_primary = True
    round_cfg.reference_phase_mix = 0.35
    round_cfg.lin2025_reference_weak_mix = 0.35
    round_runner = HybridHolographyRunner(round_cfg)
    round_phase, _ = round_runner.run_once(use_warm_start=False, save_subdir=f"round_{route_name}")

    flat_cfg = _base_cfg(flat_ref, "benchmark_feedback_proxy_flat")
    flat_cfg = _configure_lin_hybrid_route(flat_cfg, mode="quality")
    flat_cfg.feedback_target_gain = 0.08
    flat_runner = HybridHolographyRunner(flat_cfg)
    flat_runner.previous_phase = round_phase
    flat_runner.previous_target_amp = round_runner.previous_target_amp
    flat_runner.previous_target_phase = round_runner.previous_target_phase

    phase0, metrics0 = flat_runner.run_once(use_warm_start=True, save_subdir=f"flat_{route_name}_round0")
    output_dir0 = flat_cfg.output_dir / f"flat_{route_name}_round0"
    problem = flat_runner.pipeline.build_problem()
    phase_tensor = torch.tensor(np.load(output_dir0 / "slm_phase.npy"), dtype=torch.float32, device=flat_runner.pipeline.device)
    out_amp, _ = flat_runner.pipeline.propagator(problem["beam"], phase_tensor)
    measurement_file = flat_cfg.output_dir / f"{route_name}_measured.npy"
    measurement_file.parent.mkdir(parents=True, exist_ok=True)
    measured_i = out_amp.square()
    target_i = problem["target_amp"].square().to(measured_i.device)
    proxy_i = 0.60 * measured_i + 0.40 * target_i
    np.save(measurement_file, proxy_i.detach().cpu().numpy())

    flat_runner.camera_loop.enabled = True
    flat_runner.camera_loop.measured_intensity_path = str(measurement_file)
    _, metrics1 = flat_runner.run_once(use_warm_start=True, save_subdir=f"flat_{route_name}_round1")
    flat_runner.camera_loop.enabled = False
    flat_runner.camera_loop.measured_intensity_path = None
    metrics1["feedback_proxy_used"] = True
    metrics1["feedback_proxy_measurement"] = str(measurement_file)
    metrics1["feedback_proxy_measurement_blend"] = {"measured": 0.60, "target": 0.40}
    metrics1["feedback_proxy_delta_overlap"] = float(metrics1.get("overlap", 0.0) - metrics0.get("overlap", 0.0))
    metrics1["feedback_proxy_delta_efficiency"] = float(metrics1.get("efficiency", 0.0) - metrics0.get("efficiency", 0.0))
    metrics1["feedback_proxy_delta_core_uniformity_loss"] = float(metrics1.get("core_uniformity_loss", 0.0) - metrics0.get("core_uniformity_loss", 0.0))
    metrics1["feedback_proxy_delta_core_phase_flatness"] = float(metrics1.get("core_phase_flatness", 0.0) - metrics0.get("core_phase_flatness", 0.0))
    summary["routes"][route_name] = metrics1


refs = find_795_reference_files(Path("ftl_gen"))
flat_ref = _select_reference(refs, "flat_top")
round_ref = _select_reference(refs, "round_top")

summary = {"routes": {}, "notebook_version": "v82", "quality_checkpoint_subdir": QUALITY_CHECKPOINT_SUBDIR}


````

````python
# 795 only (best direct initialization without hybrid polish)
from ai_holography.references import load_reference_phase

round_cfg = _base_cfg(round_ref, "benchmark_795_round")
round_runner = HybridHolographyRunner(round_cfg)
round_problem = round_runner.pipeline.build_problem()
# Directly load reference phase instead of using _build_init_phase
round_phase = load_reference_phase(
    round_ref,
    size=(round_cfg.slm_size, round_cfg.slm_size),
    device=round_cfg.device,
)
round_runner.previous_phase = round_phase
round_runner.previous_target_amp = round_problem["target_amp"]
round_runner.previous_target_phase = round_problem["target_phase"]
_ = _evaluate_direct(round_cfg, round_phase, "round_795_only")

flat_cfg = _base_cfg(flat_ref, "benchmark_795_flat")
flat_runner = HybridHolographyRunner(flat_cfg)
# Directly load reference phase instead of using _build_init_phase
flat_phase = load_reference_phase(
    flat_ref,
    size=(flat_cfg.slm_size, flat_cfg.slm_size),
    device=flat_cfg.device,
)
summary["routes"]["795_only"] = _evaluate_direct(flat_cfg, flat_phase, "flat_795_only")

_run_reference_hybrid_route(round_ref, flat_ref, summary=summary)
_run_lin2025_hybrid_route(round_ref, flat_ref, mode="balanced", summary=summary)
_run_lin2025_hybrid_route(round_ref, flat_ref, mode="quality", summary=summary)
_run_feedback_proxy_route(round_ref, flat_ref, summary=summary)

primary_routes = {name: metrics for name, metrics in summary["routes"].items() if not metrics.get("exploratory_route", False)}
ranked = sorted(primary_routes.items(), key=lambda item: _rank_key(item[1]))
ranked_all = sorted(summary["routes"].items(), key=lambda item: _rank_key(item[1]))
summary["exploratory_routes"] = [name for name, metrics in summary["routes"].items() if metrics.get("exploratory_route", False)]
summary["ranking"] = [
    {"route": name, "rank_key": [float(value) for value in _rank_key(metrics)], "score": _route_score(metrics), "metrics": metrics}
    for name, metrics in ranked
]
summary["best_route"] = ranked[0][0] if ranked else None
summary["best_route_reason"] = (
    f"rank(core_phase, core_uniformity, overlap_loss)=({_rank_key(ranked[0][1])[0]:.6f}, {_rank_key(ranked[0][1])[1]:.6f}, {_rank_key(ranked[0][1])[2]:.6f}), diagnostic_score={_route_score(ranked[0][1]):.6f}"
    if ranked else None
)
summary["ranking_policy"] = "core_phase_flatness -> core_uniformity_loss -> overlap_loss -> diagnostic_score"

print('=== Route Summary ===')
print('ranking policy: core_phase_flatness -> core_uniformity_loss -> overlap_loss -> diagnostic_score')
print()
for name, metrics in summary["routes"].items():
    if metrics.get("exploratory_route", False):
        continue
    print(f'[{name}]')
    print(f'  overlap: {metrics.get("overlap", 0.0)}')
    print(f'  core_uniformity_loss: {metrics.get("core_uniformity_loss", 0.0)}')
    print(f'  core_phase_flatness: {metrics.get("core_phase_flatness", 0.0)}')
    print(f'  efficiency: {metrics.get("efficiency", 0.0)}')
    if 'quality_variant' in metrics:
        print(f'  quality_variant: {metrics.get("quality_variant")}')
    if 'selected_checkpoint' in metrics:
        print(f'  selected_checkpoint: {metrics.get("selected_checkpoint")}')
    print()
print('=== Ranking ===')
for i, (name, metrics) in enumerate(ranked, 1):
    print(f'{i}. {name} | score={_route_score(metrics):.6f}')
print()
if summary.get("exploratory_routes"):
    print('Exploratory routes excluded from BEST ROUTE:', ', '.join(summary["exploratory_routes"]))
    print()
print('=== Best Route ===')
if summary.get("best_route"):
    best_metrics = summary["routes"][summary["best_route"]]
    print(f'BEST ROUTE = {summary["best_route"]}')
    print(f'BEST ROUTE REASON = rank(core_phase, core_uniformity, overlap_loss)=({_rank_key(best_metrics)[0]:.6f}, {_rank_key(best_metrics)[1]:.6f}, {_rank_key(best_metrics)[2]:.6f}), diagnostic_score={_route_score(best_metrics):.6f}')
    print(f'  overlap: {best_metrics.get("overlap", 0.0)}')
    print(f'  core_uniformity_loss: {best_metrics.get("core_uniformity_loss", 0.0)}')
    print(f'  core_phase_flatness: {best_metrics.get("core_phase_flatness", 0.0)}')
    print(f'  efficiency: {best_metrics.get("efficiency", 0.0)}')
    print(f'  efficiency_floor: {best_metrics.get("efficiency_floor")}')
    print(f'  below_efficiency_floor: {best_metrics.get("below_efficiency_floor")}')

benchmark_dir = Path("ai_holography") / "runs" / "benchmark_lin2025_hybrid"
benchmark_dir.mkdir(parents=True, exist_ok=True)
summary_path = benchmark_dir / "summary.json"
summary_path.write_text(json.dumps(summary, indent=2), encoding="utf-8")
print(f"Summary saved to {summary_path}")
````

````python
import torch
import torch.nn.functional as F
import numpy as np
import math
import time
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
import scipy.optimize

torch.set_default_dtype(torch.float64)

SLM_H, SLM_W = 1024, 1272
OUT_H, OUT_W = 6876, 6876
PIXEL_PITCH = 12.5e-6
WAVELENGTH = 795e-9
FOCAL_LENGTH = 0.2

STAGE1_ITERS = 120
STAGE2_ITERS = 120
NUM_ITERS = STAGE1_ITERS + STAGE2_ITERS
STAGE1_ALPHA = 3000.0
STAGE1_LR = 0.15
STAGE2_ALPHA = 0.0
STAGE2_LR = 0.008   # v84: 0.02→0.008 (小步幅，减缓 phase 退步)
EFFICIENCY_ALPHA = 0.3
UNIFORMITY_WEIGHT = 80.0     # v83: 50→80 (与 PHASE_FLATNESS_ALPHA 均衡)
PHASE_FLATNESS_ALPHA = 80.0    # v83: 300→80 (均衡 alpha，消除 flat_rms 退步)
SPILLOVER_ALPHA_S1 = 30.0    # v83: Stage 1 spillover weight (建立聚焦约束)
SPILLOVER_ALPHA_S2 = 10.0    # v83: Stage 2 spillover weight (维持，不主导)
STAGE2_COST_SE_WEIGHT = 0.001   # v86: Stage 2 cost_se缩放增强 (0.001×5240≈5.24, 与ph_loss≈12.69约1:2.4)
PHASE_CEILING_PENALTY = 1e6    # v85: L-BFGS closure 相位上界硬惩罚系数
PRIORITY_WEIGHT_EXP = 1.5
SEED_QUICK_ITERS = 30
NUM_SEEDS_FINAL = 2
NUM_SEEDS_CANDIDATE = 3
COST_SE_LBFGS_ITERS = 30
COST_SE_LBFGS_ROUNDS = 5  # v86: progressive ceiling needs more accepted rounds
COST_SE_CEILING_GROWTH = 1.08  # v88: slower ceiling growth to keep phase near round2 balance
COST_SE_PHASE_CAP = 0.60       # v88: global cap prevents phase drifting to 0.65+
COST_SE_OVERLAP_RETENTION = 0.975  # v88: preserve high-overlap pipeline advantage
OVERLAP_FLOOR = 0.70

# === v89: Bowman-first 三阶段流水线参数 ===
V89_STAGE_A_ITERS = 50          # Stage A: 复刻 Bowman 的 fmin_cg 迭代次数
V89_STAGE_B_ITERS = 40          # Stage B: 软效率推压迭代次数
V89_STAGE_B_EFF_POWER = 2.0     # Stage B: (1/eff)^k 中的 k；2 软推压，4/8 强推压
V89_STAGE_C_ROUNDS = 3          # Stage C: LBFGS 轮数 (v88 默认 5)
V89_STAGE_C_PHASE_CAP = 0.45    # Stage C: phase cap (rad)，比 v88 (0.60) 严格 25%
V89_STAGE_C_CEILING_GROWTH = 1.05   # 比 v88 (1.08) 严格
V89_STAGE_C_OVERLAP_RETENTION = 0.985  # 比 v88 (0.975) 严格

# === v92: Stage B snapshot pool + Adam trust-region C + WGS backtracking D ===
V92_STAGE_B_ITERS = 60
V92_STAGE_B_EFF_POWER = 2.0
V92_PHASE_THRESHOLD_B = 0.30

# Stage C: replace aggressive L-BFGS with small-step Adam trust-region polish
V92_STAGE_C_STEPS = 60
V92_STAGE_C_LR = 0.0015
V92_STAGE_C_EVAL_EVERY = 5
V92_STAGE_C_SOFT_PHASE = 0.38
V92_STAGE_C_HARD_PHASE = 0.45
V92_STAGE_C_OVERLAP_RETENTION = 0.94
V92_STAGE_C_FLAT_IMPROVE = 0.95
V92_STAGE_C_GRAD_CLIP = 1.0
V92_STAGE_C_UNIFORMITY_WEIGHT = 1.0
V92_STAGE_C_COST_WEIGHT = 0.3
V92_STAGE_C_SPILL_WEIGHT = 0.1
V92_STAGE_C_PHASE_SOFT_WEIGHT = 10.0
V92_STAGE_C_PHASE_HARD_WEIGHT = 1.0e4
V92_STAGE_C_OVERLAP_WEIGHT = 100.0
V92_STAGE_C_PLATEAU_FRAC = 0.5

# Stage D: per-round WGS alpha backtracking
V92_STAGE_D_ROUNDS = 10
V92_STAGE_D_ALPHAS = [0.50, 0.35, 0.25, 0.15, 0.08, 0.04]
V92_STAGE_D_PHASE_CAP = 0.45
V92_STAGE_D_MIN_IMPROVEMENT = 0.97
V92_STAGE_D_OVERLAP_RETENTION = 0.97
V92_STAGE_D_EFF_RETENTION = 0.95
V92_STAGE_D_SPILL_MAX_GROWTH = 1.10

# === v93: Bowman warm-start + anchor + Stage C/D 锁紧 + TV + 真 Pareto ===
V93_STAGE0_ITERS = 80                # Bowman warm-start fmin_cg maxiter
V93_STAGE0_LG_SEED = 42              # LG init seed for Stage 0
V93_STAGEA_ITERS = 50                # Anchored efficiency lift fmin_cg maxiter
V93_STAGEA_EFF_POWER = 2.0           # (1/eff)^k 中的 k
V93_STAGEA_LAMBDA_INIT = 5000.0      # 锚点惩罚初始权重
V93_STAGEA_LAMBDA_MIN = 300.0        # v94: 100 → 300，永远保留约束
V93_STAGEA_LAMBDA_DECAY_EVERY = 20   # v94: 10 → 20，让 phase 收敛到 ~0.32
V93_STAGEC_SOFT_PHASE = 0.32         # 比 v92 (0.38) 严
V93_STAGEC_HARD_PHASE = 0.40         # 比 v92 (0.45) 严
V93_STAGEC_OVERLAP_RETENTION = 0.93
V93_STAGEC_TV_WEIGHT = 1.0e-5        # SLM 面 TV 正则 (Gemini 提议)
V93_STAGED_PHASE_CAP = 0.42          # v94: 0.40 → 0.42 (多 0.02 rad headroom)
V93_STAGED_ALPHAS = [0.50, 0.35, 0.25, 0.15, 0.08, 0.04, 0.02]
V93_STAGED_TV_LR = 5.0e-4            # WGS 后 TV-prox 步长
V93_STAGED_TV_STEPS = 1              # 每轮 TV-prox 迭代步数

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

focal_spx = FOCAL_LENGTH * WAVELENGTH / (PIXEL_PITCH * OUT_W)
focal_spy = FOCAL_LENGTH * WAVELENGTH / (PIXEL_PITCH * OUT_H)
print(f"系统参数: SLM={SLM_H}x{SLM_W}, Output={OUT_H}x{OUT_W}")
print(f"焦平面像素尺寸: focal_spx={focal_spx*1e6:.4f} um, focal_spy={focal_spy*1e6:.4f} um")
print(f"阶段1: {STAGE1_ITERS} iter, alpha={STAGE1_ALPHA}, lr={STAGE1_LR}")
print(f"阶段1: spill_s1={SPILLOVER_ALPHA_S1}, 阶段2: alpha={STAGE2_ALPHA}, lr={STAGE2_LR}, ph={PHASE_FLATNESS_ALPHA}, spill_s2={SPILLOVER_ALPHA_S2} (效率+相位约束)")
print(f"效率损失权重: {EFFICIENCY_ALPHA}")
print(f"智能种子选择: {NUM_SEEDS_CANDIDATE} 候选 -> top-{NUM_SEEDS_FINAL} 完整训练")


def make_beam():
    sx = sy = 3.5e-3
    sigmax = math.sqrt(2.0) * (sx / PIXEL_PITCH)
    sigmay = math.sqrt(2.0) * (sy / PIXEL_PITCH)
    y = torch.arange(SLM_H, device=device, dtype=torch.float64) - SLM_H / 2
    x = torch.arange(SLM_W, device=device, dtype=torch.float64) - SLM_W / 2
    X, Y = torch.meshgrid(x, y, indexing="xy")
    beam = torch.exp(-2.0 * ((X / sigmax) ** 2 + (Y / sigmay) ** 2))
    I_L_tot = beam.square().sum()
    beam = beam * (10000.0 / I_L_tot).sqrt()
    return beam


def propagate(field_slm):
    h, w = field_slm.shape
    oh, ow = OUT_H, OUT_W
    pad_h = (oh - h) // 2
    pad_w = (ow - w) // 2
    padded = F.pad(field_slm.unsqueeze(0).unsqueeze(0),
                   [pad_w, pad_w, pad_h, pad_h], mode="constant", value=0).squeeze()
    A0 = 1.0 / math.sqrt(oh * ow)
    output = A0 * torch.fft.ifftshift(torch.fft.fft2(torch.fft.fftshift(padded)))
    return output


def compute_overlap(field, target):
    inner = (field.conj() * target).sum().abs()
    norm = field.abs().square().sum().sqrt() * target.abs().square().sum().sqrt()
    return (inner / norm.clamp(min=1e-30)).real.item()


def compute_overlap_loss(output, target_field, norm_tgt):
    inner = (output.conj() * target_field).sum().abs()
    norm_out = output.abs().square().sum().sqrt()
    overlap = inner / (norm_out * norm_tgt).clamp(min=1e-30)
    return (1.0 - overlap) ** 2


def compute_efficiency_loss(output, beam, Wcg):
    output_amp = output.abs()
    signal_power = output_amp[Wcg > 0].square().sum()
    total_power = beam.square().sum()
    eff = signal_power / total_power.clamp(min=1e-30)
    return (1.0 - eff) ** 2


def compute_uniformity_loss(output, Wcg):
    output_amp = output.abs()
    core = output_amp[Wcg > 0]
    if core.numel() < 10:
        return torch.tensor(0.0, device=output.device)
    core_mean = core.mean()
    core_std = core.std()
    uniformity = core_std / core_mean.clamp(min=1e-12)
    return uniformity


def compute_spillover_loss(output, Wcg, beam):
    I = output.abs().square()
    total_input = beam.square().sum().clamp(min=1e-30)
    out_of_band = I[Wcg == 0].sum()
    return out_of_band / total_input


def compute_phase_flatness_loss(output, target_amp, frac=0.5):
    """v82: differentiable plateau phase flatness loss (returns variance in rad^2).

    Same calculation as ``phase_flatness_metric``, but returns a tensor for
    backprop. Intensity weights are detached: they fix the spatial weighting
    and let gradient flow only through the phase. Using ``atan2(sin, cos)`` to
    measure deviation from the circular mean keeps it ±π wrap safe."""
    flat_mask = target_amp > frac * target_amp.max()
    if int(flat_mask.sum().item()) < 10:
        return torch.tensor(0.0, device=output.device, dtype=output.real.dtype)
    amp = output.abs()
    phi = torch.angle(output)
    w = (amp * flat_mask.to(amp.dtype)).detach()
    w = w / w.sum().clamp(min=1e-30)
    cmean = (w * torch.exp(1j * phi)).sum()
    phi_ref = torch.angle(cmean)
    dphi = torch.atan2(torch.sin(phi - phi_ref), torch.cos(phi - phi_ref))
    return (w * dphi.square()).sum()


def make_priority_weight(target_amp, Wcg):
    ta_norm = target_amp / target_amp.max().clamp(min=1e-12)
    priority = (ta_norm.clamp(min=0.0) ** PRIORITY_WEIGHT_EXP + 0.1)
    return Wcg * priority


def make_target_and_weight():
    sigma = round(100e-6 / focal_spx)
    sigmax_line = round(20e-6 / focal_spy)
    sigmay_line = round(20e-6 / focal_spx)
    dx = round(-100e-6 / focal_spx)
    dy = round(-100e-6 / focal_spx)
    r0 = [dx + OUT_H / 2, dy + OUT_W / 2]
    print(f"目标参数: sigma(d)={sigma}px, sigmax={sigmax_line}px, sigmay={sigmay_line}px")
    print(f"偏移: dx={dx}px, dy={dy}px, r0={r0}")

    cols, rows = OUT_H, OUT_W
    y = torch.arange(rows, device=device, dtype=torch.float64)
    x = torch.arange(cols, device=device, dtype=torch.float64)
    X, Y = torch.meshgrid(x, y, indexing="xy")
    fx = 0.5 * (torch.abs(X - sigma / 2.0 - r0[0]) + torch.abs(X + sigma / 2.0 - r0[0]) - sigma)
    target_amp = torch.exp(-(fx ** 2 / max(sigmax_line, 1) ** 2 + (Y - r0[1]) ** 2 / max(sigmay_line, 1) ** 2))
    target_phase = torch.zeros_like(target_amp)

    wcg_radius = round(120e-6 / focal_spx)
    r = torch.sqrt((X - r0[0]) ** 2 + (Y - r0[1]) ** 2)
    flat_mask = torch.zeros((cols, rows), device=device, dtype=torch.float64)
    flat_mask[r < wcg_radius / 2] = 1.0
    Wcg = torch.ones_like(target_amp)
    threshold = 1e-4 * target_amp.max()
    Wcg[torch.abs(target_amp) < threshold] = 0.0
    Wcg = Wcg * flat_mask

    I_L_tot = make_beam().square().sum()
    target_amp = target_amp * Wcg
    target_phase = target_phase * Wcg
    I_Ta_w = target_amp.square().sum()
    target_amp = target_amp * torch.sqrt(I_L_tot / I_Ta_w.clamp(min=1e-30))

    return target_amp, target_phase, Wcg


def proper_uniformity(output_amp, target_amp, frac=0.9):
    """RMS/PV computed only on the truly flat plateau pixels of the target."""
    flat_mask = target_amp > frac * target_amp.max()
    n = int(flat_mask.sum().item())
    if n < 10:
        return float('nan'), float('nan'), n
    I = output_amp[flat_mask].square()
    mean = I.mean().clamp(min=1e-30)
    rms = (I.std() / mean).item()
    pv = ((I.max() - I.min()) / mean).item()
    return rms, pv, n


def phase_flatness_metric(output, target_amp, frac=0.5):
    """Intensity-weighted circular std of the phase across the plateau region (rad)."""
    flat_mask = target_amp > frac * target_amp.max()
    if int(flat_mask.sum().item()) < 10:
        return float('nan')
    amp = output.abs()
    phi = torch.angle(output)
    w = amp * flat_mask.to(amp.dtype)
    w_sum = w.sum().clamp(min=1e-30)
    w_n = w / w_sum
    cmean = (w_n * torch.exp(1j * phi)).sum()
    phi_ref = torch.angle(cmean)
    dphi = torch.atan2(torch.sin(phi - phi_ref), torch.cos(phi - phi_ref))
    var = (w_n * dphi.square()).sum()
    return var.sqrt().item()


def spillover_ratio(output_amp, Wcg, beam):
    I = output_amp.square()
    total_input = beam.square().sum().clamp(min=1e-30)
    in_band = I[Wcg > 0].sum() / total_input
    out_of_band = I[Wcg == 0].sum() / total_input
    return out_of_band.item(), in_band.item()


def evaluate_metrics(phase, beam, target_amp, target_phase, Wcg, priority_w=None):
    with torch.no_grad():
        field = beam * torch.exp(1j * phase)
        output = propagate(field)
        output_amp = output.abs()
        target_field = target_amp * torch.exp(1j * target_phase)

        overlap = compute_overlap(output, target_field)

        Ta_sq = target_amp.square()
        E_out_sq = output_amp.square()
        if priority_w is not None:
            cost_se = ((Ta_sq - E_out_sq) * priority_w).square().sum().item()
        else:
            cost_se = ((Ta_sq - E_out_sq) * Wcg).square().sum().item()

        target_region = target_amp > target_amp.max() * 0.5
        total_power = beam.square().sum().item()
        signal_power = output_amp[target_region].square().sum().item()
        efficiency = signal_power / max(total_power, 1e-12)

        oh, ow = output_amp.shape
        cy, cx = oh // 2, ow // 2
        roi = round(0.3 * oh)
        core = output_amp[cy-roi:cy+roi, cx-roi:cx+roi]
        core_mean = core.mean()
        core_std = core.std()
        uniformity = (core_std / core_mean.clamp(min=1e-12)).item()

        in_wcg = output_amp[Wcg > 0]
        tgt_in_wcg = target_amp[Wcg > 0]
        if in_wcg.numel() > 1:
            wcg_corr = torch.corrcoef(torch.stack([tgt_in_wcg.flatten(), in_wcg.flatten()]))[0, 1].item()
        else:
            wcg_corr = 0.0

        proper_rms, proper_pv, flat_pix = proper_uniformity(output_amp, target_amp, frac=0.9)
        phase_std = phase_flatness_metric(output, target_amp, frac=0.5)
        spill, in_band = spillover_ratio(output_amp, Wcg, beam)

        return {
            "overlap": overlap,
            "efficiency": efficiency,
            "wcg_corr": wcg_corr,
            "core_uniformity": uniformity,
            "cost_se": cost_se,
            "output_amp": output_amp.cpu(),
            "flat_rms": proper_rms,
            "flat_pv": proper_pv,
            "flat_pixels": flat_pix,
            "phase_flatness": phase_std,
            "spillover": spill,
            "in_band": in_band,
        }


def make_lg_phase(seed, curv_base=3.0, charge_spread=0.1):
    rng = torch.Generator(device=device)
    rng.manual_seed(seed)
    curv = curv_base + torch.randn(1, generator=rng, device=device).item() * 0.5
    charge = (math.pi / 4.0) + torch.randn(1, generator=rng, device=device).item() * charge_spread
    y = torch.arange(SLM_H, device=device, dtype=torch.float64) - SLM_H / 2
    x = torch.arange(SLM_W, device=device, dtype=torch.float64) - SLM_W / 2
    X, Y = torch.meshgrid(x, y, indexing="xy")
    R = torch.sqrt(X ** 2 + Y ** 2)
    Theta = torch.atan2(Y, X)
    phase = -math.pi / 2.0 + 0.5 * R + (curv / 1000.0) * R ** 2 + charge * Theta
    tilt_x = torch.randn(1, generator=rng, device=device).item() * 0.01
    tilt_y = torch.randn(1, generator=rng, device=device).item() * 0.01
    phase = phase + tilt_x * X + tilt_y * Y
    phase = torch.remainder(phase + math.pi, 2 * math.pi) - math.pi
    return phase


def run_adam_single_stage(phase_init, beam, target_amp, Wcg, target_field, norm_tgt,
                          num_iters, lr, alpha, label, use_priority=False, use_efficiency=False,
                          track_by_cost=False, use_uniformity=False, use_phase_flatness=False,
                          use_spillover=False, spillover_alpha=None, track_by_phase=False,
                          phase_ceiling_offset=0.12, cost_se_weight=1.0):
    Ta_sq = target_amp.square()
    if spillover_alpha is None:
        spillover_alpha = SPILLOVER_ALPHA_S2  # v83: default to S2 weight
    if use_priority:
        loss_weight = make_priority_weight(target_amp, Wcg)
    else:
        loss_weight = Wcg
    phase = phase_init.clone().requires_grad_(True)
    optimizer = torch.optim.Adam([phase], lr=lr)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, num_iters, eta_min=lr * 0.05)
    best_phase = phase.detach().clone()
    best_overlap = -1.0
    best_cost = float('inf')
    if track_by_phase:
        with torch.no_grad():
            init_out = propagate(beam * torch.exp(1j * phase_init))
        init_phase_flat = phase_flatness_metric(init_out, target_amp, frac=0.5)
        best_cost = float('inf')
        phase_ceiling = init_phase_flat + phase_ceiling_offset  # v83: offset=0.12 (from 0.08)
    t0 = time.time()
    for i in range(num_iters):
        optimizer.zero_grad()
        field = beam * torch.exp(1j * phase)
        output = propagate(field)
        E_out_2 = output.abs().square()
        cost_se = ((Ta_sq - E_out_2) * loss_weight).square().sum()
        overlap_loss = compute_overlap_loss(output, target_field, norm_tgt)
        cost = cost_se_weight * cost_se + alpha * overlap_loss  # v85: scaled cost_se
        if use_efficiency:
            eff_loss = compute_efficiency_loss(output, beam, Wcg)
            cost = cost + EFFICIENCY_ALPHA * eff_loss
        if use_uniformity:
            uni_loss = compute_uniformity_loss(output, Wcg)
            cost = cost + UNIFORMITY_WEIGHT * uni_loss
        if use_phase_flatness:
            ph_loss = compute_phase_flatness_loss(output, target_amp, frac=0.5)
            cost = cost + PHASE_FLATNESS_ALPHA * ph_loss
        if use_spillover:
            sp_loss = compute_spillover_loss(output, Wcg, beam)
            cost = cost + spillover_alpha * sp_loss  # v83: per-stage alpha
        cost.backward()
        optimizer.step()
        scheduler.step()
        with torch.no_grad():
            overlap = compute_overlap(output, target_field)
        if track_by_phase:
            current_phase_flat = phase_flatness_metric(output, target_amp, frac=0.5)
            if current_phase_flat <= phase_ceiling:
                current_cost = cost_se.item()
                if current_cost < best_cost:
                    best_cost = current_cost
                    best_phase = phase.detach().clone()
                    best_overlap = overlap
        elif track_by_cost:
            current_cost = cost_se.item()
            if current_cost < best_cost:
                best_cost = current_cost
                best_phase = phase.detach().clone()
                best_overlap = overlap
        else:
            if overlap > best_overlap:
                best_overlap = overlap
                best_phase = phase.detach().clone()
        if (i + 1) % 50 == 0:
            elapsed = time.time() - t0
            extras = []
            if use_efficiency:
                extras.append(f"eff_loss={EFFICIENCY_ALPHA * eff_loss.item():.1f}")
            if use_phase_flatness:
                extras.append(f"ph_loss={PHASE_FLATNESS_ALPHA * ph_loss.item():.2f}")
            extra = (", " + ", ".join(extras)) if extras else ""
            print(f"  [{label}] iter {i+1}/{num_iters}: overlap={overlap:.6f}, "
                  f"cost_se={cost_se.item():.2e}, lr={scheduler.get_last_lr()[0]:.4f}, "
                  f"time={elapsed:.1f}s{extra}")
    elapsed = time.time() - t0
    print(f"  [{label}] 完成: best_overlap={best_overlap:.6f}, 耗时={elapsed:.1f}s")
    return best_phase, best_overlap, elapsed


def run_two_stage_adam(phase_init, beam, target_amp, Wcg, target_field, norm_tgt, label="Pipe"):
    priority_w = make_priority_weight(target_amp, Wcg)

    print(f"  [{label}] 阶段1: {STAGE1_ITERS} iter, alpha={STAGE1_ALPHA}, lr={STAGE1_LR}, spill_s1={SPILLOVER_ALPHA_S1} (overlap优先+spillover约束)")
    phase_s1, overlap_s1, time_s1 = run_adam_single_stage(
        phase_init, beam, target_amp, Wcg, target_field, norm_tgt,
        num_iters=STAGE1_ITERS, lr=STAGE1_LR, alpha=STAGE1_ALPHA,
        label=f"{label} S1", use_priority=False, use_efficiency=False,
        use_spillover=True, spillover_alpha=SPILLOVER_ALPHA_S1)  # v83: spillover in S1

    print(f"  [{label}] 阶段2: {STAGE2_ITERS} iter, alpha={STAGE2_ALPHA}, lr={STAGE2_LR}, ph={PHASE_FLATNESS_ALPHA}, spill_s2={SPILLOVER_ALPHA_S2} (效率+相位约束,ceiling+0.12)")
    phase_s2, overlap_s2, time_s2 = run_adam_single_stage(
        phase_s1, beam, target_amp, Wcg, target_field, norm_tgt,
        num_iters=STAGE2_ITERS, lr=STAGE2_LR, alpha=STAGE2_ALPHA,
        label=f"{label} S2", use_priority=True, use_efficiency=True,
        use_uniformity=True, use_phase_flatness=True,
        use_spillover=True, spillover_alpha=SPILLOVER_ALPHA_S2,
        track_by_phase=True, phase_ceiling_offset=0.12,
        cost_se_weight=STAGE2_COST_SE_WEIGHT)  # v85: 0.0002*cost_se, phase梯度主导(94%)

    with torch.no_grad():
        Ta_sq = target_amp.square()
        for pname, pphase in [("S1", phase_s1), ("S2", phase_s2)]:
            field = beam * torch.exp(1j * pphase)
            output = propagate(field)
            cost = ((Ta_sq - output.abs().square()) * priority_w).square().sum().item()
            ph = phase_flatness_metric(output, target_amp, frac=0.5)
            ov = compute_overlap(output, target_field)
            print(f"  [{label}] {pname}结果: overlap={ov:.6f}, cost_se={cost:.2e}, phase_flat={ph:.3f}rad")

    return phase_s1, overlap_s1, time_s1, phase_s2, overlap_s2, time_s2

def select_pareto_candidate(phases_data, beam, target_amp, target_phase, Wcg, priority_w):
    """v82: 5-dim Rank-Pareto over {overlap, efficiency, cost_se, phase_flatness, flat_rms}.

    Each rank is 0-indexed where 0 = best in that dimension. The combined score
    is a weighted sum of ranks; we pick the candidate with the *lowest* score.
    v83 weights: overlap=1, efficiency=1, cost_se=1, phase_flatness=3, flat_rms=2.
    (v82 had cost_se=3, phase=2, rms=1 — cost_se dominated after phase loss drove
    it 4x higher, making seed selection pick low-cost/high-phase candidates.)
    Fixes a sign bug in the v75/v76 version."""
    raw = []
    for name, phase, overlap, time_s in phases_data:
        m = evaluate_metrics(phase, beam, target_amp, target_phase, Wcg, priority_w)
        raw.append((name, phase, m, time_s))
        rms_s = f"{m['flat_rms']:.3f}" if not (isinstance(m['flat_rms'], float) and math.isnan(m['flat_rms'])) else "nan"
        ph_s = f"{m['phase_flatness']:.3f}" if not (isinstance(m['phase_flatness'], float) and math.isnan(m['phase_flatness'])) else "nan"
        print(f"  [{name}] overlap={m['overlap']:.6f}, eff={m['efficiency']:.4f}, "
              f"cost_se={m['cost_se']:.2e}, flat_rms={rms_s}, phase={ph_s}rad")
    n = len(raw)

    def _safe(k, key, asc):
        v = raw[k][2][key]
        if isinstance(v, float) and math.isnan(v):
            return float('inf')
        return v if asc else -v

    ranks_ov = sorted(range(n), key=lambda k: _safe(k, "overlap", asc=False))
    ranks_eff = sorted(range(n), key=lambda k: _safe(k, "efficiency", asc=False))
    ranks_cs = sorted(range(n), key=lambda k: _safe(k, "cost_se", asc=True))
    ranks_phase = sorted(range(n), key=lambda k: _safe(k, "phase_flatness", asc=True))
    ranks_rms = sorted(range(n), key=lambda k: _safe(k, "flat_rms", asc=True))

    rank_val = [0.0] * n
    for r, k in enumerate(ranks_ov):
        rank_val[k] += 1.0 * r
    for r, k in enumerate(ranks_eff):
        rank_val[k] += 1.0 * r
    for r, k in enumerate(ranks_cs):
        rank_val[k] += 1.0 * r    # v83: 3→1 (cost_se 非线性劣化后不应主导)
    for r, k in enumerate(ranks_phase):
        rank_val[k] += 3.0 * r    # v83: 2→3 (phase 是核心优化目标)
    for r, k in enumerate(ranks_rms):
        rank_val[k] += 2.0 * r    # v83: 1→2 (均匀性需要保护)

    best_idx = min(range(n), key=lambda k: rank_val[k])
    best = raw[best_idx] + (rank_val[best_idx],)
    print(f"  Rank-Pareto 最优: {best[0]} (rank_score={best[4]:.2f})")
    return best
def run_adam_train(phase_init, beam, target_amp, Wcg, target_field, norm_tgt,
                   num_iters=200, lr=0.3, label="Adam"):
    Ta_sq = target_amp.square()
    phase = phase_init.clone().requires_grad_(True)
    optimizer = torch.optim.Adam([phase], lr=lr)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, num_iters, eta_min=0.01)
    best_phase = phase.detach().clone()
    best_overlap = -1.0
    t0 = time.time()
    for i in range(num_iters):
        optimizer.zero_grad()
        field = beam * torch.exp(1j * phase)
        output = propagate(field)
        E_out_2 = output.abs().square()
        cost = ((Ta_sq - E_out_2) * Wcg).square().sum()
        cost.backward()
        optimizer.step()
        scheduler.step()
        with torch.no_grad():
            overlap = compute_overlap(output, target_field)
            if overlap > best_overlap:
                best_overlap = overlap
                best_phase = phase.detach().clone()
        if (i + 1) % 50 == 0:
            elapsed = time.time() - t0
            print(f"  [{label}] iter {i+1}/{num_iters}: overlap={overlap:.6f}, "
                  f"cost_se={cost.item():.2e}, lr={scheduler.get_last_lr()[0]:.4f}, "
                  f"time={elapsed:.1f}s")
    elapsed = time.time() - t0
    print(f"  [{label}] 完成: best_overlap={best_overlap:.6f}, 耗时={elapsed:.1f}s")
    return best_phase, elapsed


def run_lbfgs_polish(phase_init, beam, target_amp, Wcg, target_field,
                     rounds=3, max_iter=10, lr=0.01, label="L-BFGS"):
    Ta_sq = target_amp.square()
    phase = phase_init.clone().requires_grad_(True)
    best_phase = phase_init.clone()
    with torch.no_grad():
        field = beam * torch.exp(1j * phase_init)
        output = propagate(field)
        best_overlap = compute_overlap(output, target_field)
        init_phase = phase_flatness_metric(output, target_amp, frac=0.5)
    total_time = 0
    for rd in range(rounds):
        optimizer_lb = torch.optim.LBFGS([phase], lr=lr, max_iter=max_iter,
                                         history_size=10, line_search_fn="strong_wolfe")
        t1 = time.time()
        def closure():
            optimizer_lb.zero_grad()
            field = beam * torch.exp(1j * phase)
            output = propagate(field)
            E_out_2 = output.abs().square()
            cost = ((Ta_sq - E_out_2) * Wcg).square().sum()
            # v85: hard ceiling penalty — prevent overlap polish from drifting phase
            ph_var = compute_phase_flatness_loss(output, target_amp, frac=0.5)
            ceiling_var_lb = torch.tensor((init_phase * 1.08) ** 2,
                                          device=device, dtype=torch.float64)
            cost = cost + PHASE_CEILING_PENALTY * torch.relu(ph_var - ceiling_var_lb)
            cost.backward()
            return cost
        optimizer_lb.step(closure)
        round_time = time.time() - t1
        total_time += round_time
        with torch.no_grad():
            field = beam * torch.exp(1j * phase)
            output = propagate(field)
            overlap = compute_overlap(output, target_field)
            current_phase = phase_flatness_metric(output, target_amp, frac=0.5)
            if overlap > best_overlap and current_phase <= init_phase + 0.12:  # v83: 0.08→0.12
                best_overlap = overlap
                best_phase = phase.detach().clone()
                print(f"  [{label}] round {rd+1}/{rounds}: overlap={overlap:.6f}, ph={current_phase:.3f}rad, time={round_time:.1f}s")
            else:
                reason = "degraded" if overlap <= best_overlap else f"phase ({current_phase:.3f} > {init_phase+0.12:.3f})"
                print(f"  [{label}] round {rd+1}/{rounds}: overlap={overlap:.6f}, ph={current_phase:.3f}rad ({reason}, stop)")
                break
    print(f"  [{label}] 完成: best_overlap={best_overlap:.6f}, 耗时={total_time:.1f}s")
    return best_phase, total_time


def run_cost_se_lbfgs(phase_init, beam, target_amp, Wcg, target_field, label="CostSE-LBFGS"):
    """v88: balanced CostSE-LBFGS with capped progressive ceiling,
    stricter overlap retention, and rollback for rejected candidates."""
    Ta_sq = target_amp.square()
    priority_w = make_priority_weight(target_amp, Wcg)
    phase = phase_init.clone().requires_grad_(True)
    with torch.no_grad():
        field = beam * torch.exp(1j * phase_init)
        output = propagate(field)
        init_overlap = compute_overlap(output, target_field)
        init_cost = ((Ta_sq - output.abs().square()) * priority_w).square().sum().item()
        init_phase_flat = phase_flatness_metric(output, target_amp, frac=0.5)
    t0 = time.time()
    best_phase = phase.clone()
    best_cost = init_cost
    best_overlap = init_overlap
    best_phase_flat = init_phase_flat
    print(f"  [{label}] init: cost_se={init_cost:.2e}, overlap={init_overlap:.6f}, phase_flat={init_phase_flat:.3f}rad")
    for rd in range(COST_SE_LBFGS_ROUNDS):
        current_ceiling_phase = best_phase_flat
        ceiling_phase_limit = min(current_ceiling_phase * COST_SE_CEILING_GROWTH,
                                  COST_SE_PHASE_CAP)
        ceiling_var_cs = torch.tensor(ceiling_phase_limit ** 2,
                                      device=device, dtype=torch.float64)
        optimizer_lb = torch.optim.LBFGS([phase], lr=0.005, max_iter=COST_SE_LBFGS_ITERS,
                                         history_size=10, line_search_fn="strong_wolfe")
        def closure():
            optimizer_lb.zero_grad()
            field = beam * torch.exp(1j * phase)
            output = propagate(field)
            E_out_2 = output.abs().square()
            cost = ((Ta_sq - E_out_2) * priority_w).square().sum()
            # v86: progressive ceiling, captured once per round from last accepted phase_flat
            ph_var = compute_phase_flatness_loss(output, target_amp, frac=0.5)
            cost = cost + PHASE_CEILING_PENALTY * torch.relu(ph_var - ceiling_var_cs)
            cost = cost + SPILLOVER_ALPHA_S2 * compute_spillover_loss(output, Wcg, beam)
            cost.backward()
            return cost
        optimizer_lb.step(closure)
        with torch.no_grad():
            field = beam * torch.exp(1j * phase)
            output = propagate(field)
            overlap = compute_overlap(output, target_field)
            cost_val = ((Ta_sq - output.abs().square()) * priority_w).square().sum().item()
            phase_flat = phase_flatness_metric(output, target_amp, frac=0.5)
            phase_vs_ceiling = phase_flat / max(ceiling_phase_limit, 1e-12)
            print(f"  [{label}] round {rd+1}/{COST_SE_LBFGS_ROUNDS}: ceiling={ceiling_phase_limit:.3f}rad, cost_se={cost_val:.2e}, overlap={overlap:.6f}, phase_flat={phase_flat:.3f}rad, phase/ceiling={phase_vs_ceiling:.3f}")
            cost_improvement = (best_cost - cost_val) / max(best_cost, 1e-30)
            overlap_retention = overlap / max(init_overlap, 1e-30)
            phase_ok = phase_flat <= ceiling_phase_limit * 1.01
            overlap_ok = overlap_retention >= COST_SE_OVERLAP_RETENTION
            if cost_improvement > 0.10 and overlap_ok and phase_ok:
                best_cost = cost_val
                best_phase = phase.detach().clone()
                best_overlap = overlap
                best_phase_flat = phase_flat
                print(f"    Accepted: cost {cost_improvement*100:+.1f}%, overlap ret {overlap_retention*100:.1f}%, phase/ceiling {phase_vs_ceiling*100:.1f}%")
            else:
                phase.copy_(best_phase)
                print(f"    Rejected: cost {cost_improvement*100:+.1f}%, overlap ret {overlap_retention*100:.1f}%, overlap_ok={overlap_ok}, phase_ok={phase_ok}; rollback to best phase")
    elapsed = time.time() - t0
    improvement = (init_cost - best_cost) / max(init_cost, 1e-30) * 100
    print(f"  [{label}] 完成: cost_se {init_cost:.2e} -> {best_cost:.2e} ({improvement:+.1f}%), "
          f"overlap {init_overlap:.6f} -> {best_overlap:.6f}, "
          f"phase {init_phase_flat:.3f} -> {best_phase_flat:.3f}rad, 耗时={elapsed:.1f}s")
    return best_phase, elapsed


def run_795_cg(beam, target_amp, Wcg, target_field, norm_tgt):
    print(f"\n--- 795 CG (random init, Adam lr=0.3 cosine, {NUM_ITERS} iters + L-BFGS) ---")
    torch.manual_seed(42)
    phase_init = (torch.rand(SLM_H, SLM_W, device=device, dtype=torch.float64) - 0.5) * 0.1
    best_phase, adam_time = run_adam_train(
        phase_init, beam, target_amp, Wcg, target_field, norm_tgt, label="795 Adam")
    best_phase, lbfgs_time = run_lbfgs_polish(
        best_phase, beam, target_amp, Wcg, target_field, label="795 L-BFGS")
    return best_phase, adam_time + lbfgs_time


def run_bowman_cg(beam, target_amp, Wcg, target_field, norm_tgt):
    print(f"\n--- Bowman CG (L-G init, scipy fmin_cg, {NUM_ITERS} maxiter, 无 scale) ---")
    Ta_sq = target_amp.square()
    init_phase = make_lg_phase(42)
    phase_np = init_phase.detach().cpu().numpy().flatten().copy()
    eval_count = [0]
    best_overlap = [-1.0]
    best_phase_arr = [phase_np.copy()]

    def cost_and_grad(phi_flat):
        phase_t = torch.from_numpy(phi_flat).to(device=device, dtype=torch.float64).reshape(SLM_H, SLM_W)
        phase_t = phase_t.requires_grad_(True)
        field = beam * torch.exp(1j * phase_t)
        output = propagate(field)
        E_out_2 = output.abs().square()
        cost = ((Ta_sq - E_out_2) * Wcg).square().sum()
        cost.backward()
        grad = phase_t.grad.detach().cpu().numpy().flatten()
        eval_count[0] += 1
        with torch.no_grad():
            overlap = compute_overlap(output, target_field)
            if overlap > best_overlap[0]:
                best_overlap[0] = overlap
                best_phase_arr[0] = phi_flat.copy()
        if eval_count[0] % 30 == 0:
            print(f"  [Bowman CG] eval {eval_count[0]}: overlap={overlap:.6f}, cost={cost.item():.2e}")
        return cost.item(), grad

    _cache = [None, None, None]
    def cost_fn(p):
        if _cache[0] is None or not np.array_equal(p, _cache[0]):
            _cache[0] = p.copy()
            _cache[1], _cache[2] = cost_and_grad(p)
        return _cache[1]
    def grad_fn(p):
        if _cache[0] is None or not np.array_equal(p, _cache[0]):
            _cache[0] = p.copy()
            _cache[1], _cache[2] = cost_and_grad(p)
        return _cache[2]

    t0 = time.time()
    res = scipy.optimize.fmin_cg(
        f=cost_fn,
        x0=phase_np,
        fprime=grad_fn,
        maxiter=NUM_ITERS,
        disp=False,
    )
    total_time = time.time() - t0
    final_phase = torch.from_numpy(best_phase_arr[0]).to(device=device, dtype=torch.float64).reshape(SLM_H, SLM_W)
    print(f"  [Bowman CG] 完成: best_overlap={best_overlap[0]:.6f}, evals={eval_count[0]}, 耗时={total_time:.1f}s")
    return final_phase, total_time




def _v89_make_cached_cost_grad(cost_and_grad_fn):
    """Cache the latest (cost, grad) pair to avoid double evaluation between f and fprime."""
    cache = [None, None, None]

    def cost_fn(p):
        if cache[0] is None or not np.array_equal(p, cache[0]):
            cache[0] = p.copy()
            cache[1], cache[2] = cost_and_grad_fn(p)
        return cache[1]

    def grad_fn(p):
        if cache[0] is None or not np.array_equal(p, cache[0]):
            cache[0] = p.copy()
            cache[1], cache[2] = cost_and_grad_fn(p)
        return cache[2]

    return cost_fn, grad_fn


def run_v89_stage_a(beam, target_amp, Wcg, target_field, lg_seed=42):
    """Stage A: LG init + scipy fmin_cg on cost_SE, locks phase ~0.30 rad."""
    print(f"\n  [v89-A] LG init + fmin_cg on ((Ta^2 - E^2)*Wcg)^2, {V89_STAGE_A_ITERS} iter (phase lock)")
    Ta_sq = target_amp.square()
    init_phase = make_lg_phase(lg_seed)
    phase_np = init_phase.detach().cpu().numpy().flatten().copy()
    eval_count = [0]
    best = {"phase": phase_np.copy(), "cost": float("inf")}

    def cost_and_grad(phi_flat):
        phase_t = torch.from_numpy(phi_flat).to(device=device, dtype=torch.float64).reshape(SLM_H, SLM_W)
        phase_t = phase_t.requires_grad_(True)
        field = beam * torch.exp(1j * phase_t)
        output = propagate(field)
        E_out_2 = output.abs().square()
        cost = ((Ta_sq - E_out_2) * Wcg).square().sum()
        cost.backward()
        grad = phase_t.grad.detach().cpu().numpy().flatten()
        eval_count[0] += 1
        cv = cost.item()
        if cv < best["cost"]:
            best["cost"] = cv
            best["phase"] = phi_flat.copy()
        if eval_count[0] % 20 == 0:
            with torch.no_grad():
                ov = compute_overlap(output, target_field)
                ph = phase_flatness_metric(output, target_amp, frac=0.5)
            print(f"    [A] eval {eval_count[0]}: cost={cv:.2e}, overlap={ov:.4f}, phase_flat={ph:.3f}rad")
        return cv, grad

    cost_fn, grad_fn = _v89_make_cached_cost_grad(cost_and_grad)
    t0 = time.time()
    try:
        scipy.optimize.fmin_cg(f=cost_fn, x0=phase_np, fprime=grad_fn,
                               maxiter=V89_STAGE_A_ITERS, disp=False)
    except Exception as e:
        print(f"    [A] fmin_cg early stop: {e}")
    elapsed = time.time() - t0
    phase_a = torch.from_numpy(best["phase"]).to(device=device, dtype=torch.float64).reshape(SLM_H, SLM_W)
    with torch.no_grad():
        out = propagate(beam * torch.exp(1j * phase_a))
        ov = compute_overlap(out, target_field)
        ph = phase_flatness_metric(out, target_amp, frac=0.5)
        eff = (out.abs().square()[Wcg > 0].sum() / beam.square().sum()).item()
    print(f"  [v89-A] 完成: overlap={ov:.4f}, phase_flat={ph:.3f}rad, eff={eff:.4f}, evals={eval_count[0]}, 耗时={elapsed:.1f}s")
    return phase_a, elapsed


def run_v89_stage_b(phase_in, beam, target_amp, Wcg, target_field, norm_tgt):
    """Stage B: fmin_cg on (1-overlap)^2 * (1/eff)^k, mild efficiency push."""
    print(f"\n  [v89-B] fmin_cg on (1-overlap)^2 * (1/eff)^{V89_STAGE_B_EFF_POWER}, {V89_STAGE_B_ITERS} iter (efficiency lift)")
    phase_np = phase_in.detach().cpu().numpy().flatten().copy()
    total_input = beam.square().sum()
    eval_count = [0]
    # Track the candidate that best preserves phase while improving efficiency.
    # We use efficiency-weighted overlap as the objective; rely on caller to
    # validate the result (phase_flat) post-hoc.
    best = {"phase": phase_np.copy(), "cost": float("inf")}

    def cost_and_grad(phi_flat):
        phase_t = torch.from_numpy(phi_flat).to(device=device, dtype=torch.float64).reshape(SLM_H, SLM_W)
        phase_t = phase_t.requires_grad_(True)
        field = beam * torch.exp(1j * phase_t)
        output = propagate(field)
        E_out_2 = output.abs().square()
        # overlap (mirrors compute_overlap_loss but differentiable)
        inner = (output.conj() * target_field).sum().abs()
        norm_out = output.abs().square().sum().sqrt()
        overlap = inner / (norm_out * norm_tgt).clamp(min=1e-30)
        # efficiency over Wcg region
        eff = (E_out_2[Wcg > 0].sum() / total_input.clamp(min=1e-30)).clamp(min=1e-6)
        cost = 1e9 * (1.0 - overlap).square() * torch.pow(1.0 / eff, V89_STAGE_B_EFF_POWER)
        cost.backward()
        grad = phase_t.grad.detach().cpu().numpy().flatten()
        eval_count[0] += 1
        cv = cost.item()
        if cv < best["cost"]:
            best["cost"] = cv
            best["phase"] = phi_flat.copy()
        if eval_count[0] % 15 == 0:
            with torch.no_grad():
                ph = phase_flatness_metric(output, target_amp, frac=0.5)
            print(f"    [B] eval {eval_count[0]}: cost={cv:.2e}, overlap={overlap.item():.4f}, eff={eff.item():.4f}, phase_flat={ph:.3f}rad")
        return cv, grad

    cost_fn, grad_fn = _v89_make_cached_cost_grad(cost_and_grad)
    t0 = time.time()
    try:
        scipy.optimize.fmin_cg(f=cost_fn, x0=phase_np, fprime=grad_fn,
                               maxiter=V89_STAGE_B_ITERS, disp=False)
    except Exception as e:
        print(f"    [B] fmin_cg early stop: {e}")
    elapsed = time.time() - t0
    if best_constrained["phase"] is not None:
        selected_b = best_constrained
        print(f"    [B] selected constrained snapshot: cost={selected_b['cost']:.2e}, phase_flat={selected_b['phase_flat']:.3f}rad <= {V91_PHASE_THRESHOLD_B:.2f}")
    else:
        selected_b = best
        print(f"    [B] no snapshot met phase <= {V91_PHASE_THRESHOLD_B:.2f}; fallback to min cost")
    phase_b = torch.from_numpy(selected_b["phase"]).to(device=device, dtype=torch.float64).reshape(SLM_H, SLM_W)
    with torch.no_grad():
        out = propagate(beam * torch.exp(1j * phase_b))
        ov = compute_overlap(out, target_field)
        ph = phase_flatness_metric(out, target_amp, frac=0.5)
        eff = (out.abs().square()[Wcg > 0].sum() / beam.square().sum()).item()
    print(f"  [v89-B] 完成: overlap={ov:.4f}, phase_flat={ph:.3f}rad, eff={eff:.4f}, evals={eval_count[0]}, 耗时={elapsed:.1f}s")
    return phase_b, elapsed


def run_v89_stage_c(phase_in, beam, target_amp, Wcg, target_field):
    """Stage C: CostSE-LBFGS with v89 tightened phase cap (mirrors run_cost_se_lbfgs body)."""
    print(f"\n  [v89-C] CostSE-LBFGS uniformity polish (cap={V89_STAGE_C_PHASE_CAP}rad, growth={V89_STAGE_C_CEILING_GROWTH}, retention={V89_STAGE_C_OVERLAP_RETENTION})")
    Ta_sq = target_amp.square()
    priority_w = make_priority_weight(target_amp, Wcg)
    phase = phase_in.clone().requires_grad_(True)
    with torch.no_grad():
        field = beam * torch.exp(1j * phase_in)
        output = propagate(field)
        init_overlap = compute_overlap(output, target_field)
        init_cost = ((Ta_sq - output.abs().square()) * priority_w).square().sum().item()
        init_phase_flat = phase_flatness_metric(output, target_amp, frac=0.5)
    t0 = time.time()
    best_phase = phase.detach().clone()
    best_cost = init_cost
    best_overlap = init_overlap
    best_phase_flat = init_phase_flat
    print(f"  [v89-C] init: cost_se={init_cost:.2e}, overlap={init_overlap:.6f}, phase_flat={init_phase_flat:.3f}rad")
    for rd in range(V89_STAGE_C_ROUNDS):
        ceiling_phase_limit = min(best_phase_flat * V89_STAGE_C_CEILING_GROWTH,
                                  V89_STAGE_C_PHASE_CAP)
        ceiling_var_cs = torch.tensor(ceiling_phase_limit ** 2,
                                      device=device, dtype=torch.float64)
        optimizer_lb = torch.optim.LBFGS([phase], lr=0.005, max_iter=COST_SE_LBFGS_ITERS,
                                         history_size=10, line_search_fn="strong_wolfe")

        def closure():
            optimizer_lb.zero_grad()
            field = beam * torch.exp(1j * phase)
            output = propagate(field)
            E_out_2 = output.abs().square()
            cost = ((Ta_sq - E_out_2) * priority_w).square().sum()
            ph_var = compute_phase_flatness_loss(output, target_amp, frac=0.5)
            cost = cost + PHASE_CEILING_PENALTY * torch.relu(ph_var - ceiling_var_cs)
            cost = cost + SPILLOVER_ALPHA_S2 * compute_spillover_loss(output, Wcg, beam)
            cost.backward()
            return cost

        optimizer_lb.step(closure)
        with torch.no_grad():
            field = beam * torch.exp(1j * phase)
            output = propagate(field)
            overlap = compute_overlap(output, target_field)
            cost_val = ((Ta_sq - output.abs().square()) * priority_w).square().sum().item()
            phase_flat = phase_flatness_metric(output, target_amp, frac=0.5)
            phase_vs_ceiling = phase_flat / max(ceiling_phase_limit, 1e-12)
            print(f"  [v89-C] round {rd+1}/{V89_STAGE_C_ROUNDS}: ceiling={ceiling_phase_limit:.3f}rad, cost_se={cost_val:.2e}, overlap={overlap:.6f}, phase_flat={phase_flat:.3f}rad ({phase_vs_ceiling*100:.0f}% of ceiling)")
            cost_improvement = (best_cost - cost_val) / max(best_cost, 1e-30)
            overlap_retention = overlap / max(init_overlap, 1e-30)
            phase_ok = phase_flat <= ceiling_phase_limit * 1.01
            overlap_ok = overlap_retention >= V89_STAGE_C_OVERLAP_RETENTION
            if cost_improvement > 0.05 and overlap_ok and phase_ok:
                best_cost = cost_val
                best_phase = phase.detach().clone()
                best_overlap = overlap
                best_phase_flat = phase_flat
                print(f"    Accepted: cost {cost_improvement*100:+.1f}%, overlap ret {overlap_retention*100:.1f}%")
            else:
                phase.copy_(best_phase)
                print(f"    Rejected: cost {cost_improvement*100:+.1f}%, overlap ret {overlap_retention*100:.1f}%, overlap_ok={overlap_ok}, phase_ok={phase_ok}; rollback")
    elapsed = time.time() - t0
    improvement = (init_cost - best_cost) / max(init_cost, 1e-30) * 100
    print(f"  [v89-C] 完成: cost_se {init_cost:.2e} -> {best_cost:.2e} ({improvement:+.1f}%), "
          f"overlap {init_overlap:.6f} -> {best_overlap:.6f}, "
          f"phase {init_phase_flat:.3f} -> {best_phase_flat:.3f}rad, 耗时={elapsed:.1f}s")
    return best_phase, elapsed


def run_pipeline_v89(beam, target_amp, Wcg, target_field, norm_tgt):
    """v89 Pipeline: Bowman-first three-stage flow."""
    print(f"\n--- v89 Pipeline (Bowman-first: phase-lock -> eff-boost -> uniformity-polish) ---")
    phase_a, time_a = run_v89_stage_a(beam, target_amp, Wcg, target_field, lg_seed=42)
    phase_b, time_b = run_v89_stage_b(phase_a, beam, target_amp, Wcg, target_field, norm_tgt)
    phase_c, time_c = run_v89_stage_c(phase_b, beam, target_amp, Wcg, target_field)
    total = time_a + time_b + time_c
    print(f"\n  [v89] 累计耗时: A={time_a:.1f}s + B={time_b:.1f}s + C={time_c:.1f}s = {total:.1f}s")
    return phase_c, total




def _v92_value_is_finite(value):
    return not (isinstance(value, float) and math.isnan(value))


def _v92_candidate_metrics(phase, beam, target_amp, target_phase, Wcg, priority_w):
    metrics = evaluate_metrics(phase, beam, target_amp, target_phase, Wcg, priority_w)
    metrics.pop("output_amp", None)
    return metrics


def _v92_print_candidate_table(candidates):
    print("\n  [v92] Pipeline candidate pool:")
    print(f"  {'name':<24} {'overlap':>8} {'eff':>8} {'phase':>8} {'flat_rms':>9} {'cost_se':>10} {'spill':>8}")
    for c in candidates:
        m = c["metrics"]
        print(f"  {c['name']:<24} {m['overlap']:>8.4f} {m['efficiency']:>8.4f} "
              f"{m['phase_flatness']:>8.3f} {m['flat_rms']:>9.3f} {m['cost_se']:>10.2e} {m['spillover']:>8.3f}")


def _v92_select_representatives(candidates):
    def valid_value(c, key):
        v = c["metrics"][key]
        return not (isinstance(v, float) and math.isnan(v))

    phase_pool = [c for c in candidates if c["metrics"]["overlap"] >= 0.72 and valid_value(c, "phase_flatness")]
    phase_safe = min(phase_pool or candidates, key=lambda c: c["metrics"]["phase_flatness"])

    flat_pool = [c for c in candidates if valid_value(c, "flat_rms") and c["metrics"]["phase_flatness"] <= 0.55]
    flat_safe = min(flat_pool or candidates, key=lambda c: c["metrics"]["flat_rms"] if valid_value(c, "flat_rms") else float("inf"))

    cost_pool = [c for c in candidates if c["metrics"]["overlap"] >= 0.68 and c["metrics"]["phase_flatness"] <= 0.60]
    cost_safe = min(cost_pool or candidates, key=lambda c: c["metrics"]["cost_se"])

    def balanced_score(c):
        m = c["metrics"]
        flat = m["flat_rms"] if valid_value(c, "flat_rms") else 3.0
        return (
            1.3 * min(flat / 0.40, 4.0)
            + 1.4 * min(m["phase_flatness"] / 0.45, 4.0)
            + 1.0 * max(0.0, 0.74 - m["overlap"]) / 0.08
            + 0.8 * max(0.0, 0.72 - m["efficiency"]) / 0.10
            + 0.3 * min(m["spillover"] / 0.30, 4.0)
            + 0.2 * min(m["cost_se"] / 15000.0, 4.0)
        )

    balanced_pool = [c for c in candidates if c["metrics"]["overlap"] >= 0.70 and c["metrics"]["phase_flatness"] <= 0.55]
    balanced = min(balanced_pool or candidates, key=balanced_score)
    return {
        "Pipeline_phase_safe": phase_safe,
        "Pipeline_balanced": balanced,
        "Pipeline_flat_safe": flat_safe,
        "Pipeline_cost_safe": cost_safe,
    }


def _v92_print_representatives(reps):
    print("\n  [v92] Representative candidates:")
    for label, cand in reps.items():
        m = cand["metrics"]
        print(f"  {label:<22} <- {cand['name']:<20} overlap={m['overlap']:.4f}, eff={m['efficiency']:.4f}, "
              f"phase={m['phase_flatness']:.3f}, flat_rms={m['flat_rms']:.3f}, cost_se={m['cost_se']:.2e}")


def run_v92_stage_c(phase_in, beam, target_amp, Wcg, target_field, norm_tgt):
    """v92 Stage C: small-step Adam trust-region polish."""
    print(f"\n  [v92-C] Adam trust-region polish (steps={V92_STAGE_C_STEPS}, lr={V92_STAGE_C_LR}, "
          f"soft={V92_STAGE_C_SOFT_PHASE}, hard={V92_STAGE_C_HARD_PHASE}, "
          f"retention={V92_STAGE_C_OVERLAP_RETENTION})")
    Ta_sq = target_amp.square()
    plateau_mask = (target_amp > V92_STAGE_C_PLATEAU_FRAC * target_amp.max()).to(target_amp.dtype) * (Wcg > 0).to(target_amp.dtype)
    plateau_count = int(plateau_mask.sum().item())
    if plateau_count < 10:
        print(f"  [v92-C] plateau too small ({plateau_count} px), fall back to Wcg")
        plateau_mask = (Wcg > 0).to(target_amp.dtype)
        plateau_count = int(plateau_mask.sum().item())
    print(f"  [v92-C] plateau pixels: {plateau_count}")

    tgt_in = Ta_sq[plateau_mask > 0]
    tgt_norm_const = (tgt_in / tgt_in.mean().clamp(min=1e-30)).detach()

    phase = phase_in.detach().clone().requires_grad_(True)
    optimizer = torch.optim.Adam([phase], lr=V92_STAGE_C_LR)
    soft_var = torch.tensor(V92_STAGE_C_SOFT_PHASE ** 2, device=device, dtype=torch.float64)
    hard_var = torch.tensor(V92_STAGE_C_HARD_PHASE ** 2, device=device, dtype=torch.float64)

    with torch.no_grad():
        output0 = propagate(beam * torch.exp(1j * phase_in))
        init_overlap = compute_overlap(output0, target_field)
        init_cost_se = ((Ta_sq - output0.abs().square()) * plateau_mask).square().sum().item()
        init_phase_flat = phase_flatness_metric(output0, target_amp, frac=0.5)
        init_flat_rms, _, _ = proper_uniformity(output0.abs(), target_amp, frac=0.9)
        init_spill, _ = spillover_ratio(output0.abs(), Wcg, beam)

    best_phase = phase_in.detach().clone()
    best_flat_rms = init_flat_rms
    best_overlap = init_overlap
    best_phase_flat = init_phase_flat
    best_cost_se = init_cost_se
    phase_candidates = [("C_init", best_phase.clone())]
    t0 = time.time()
    print(f"  [v92-C] init: cost_se={init_cost_se:.2e}, overlap={init_overlap:.4f}, "
          f"phase_flat={init_phase_flat:.3f}rad, flat_rms={init_flat_rms:.3f}, spill={init_spill:.3f}")

    init_cost_scale = max(init_cost_se, 1e-30)
    for step in range(1, V92_STAGE_C_STEPS + 1):
        optimizer.zero_grad()
        output = propagate(beam * torch.exp(1j * phase))
        E_out_2 = output.abs().square()
        cost_se = ((Ta_sq - E_out_2) * plateau_mask).square().sum()
        out_in = E_out_2[plateau_mask > 0]
        out_norm = out_in / out_in.mean().clamp(min=1e-30)
        uni_loss = (out_norm - tgt_norm_const).square().mean()
        ph_var = compute_phase_flatness_loss(output, target_amp, frac=0.5)
        spill = compute_spillover_loss(output, Wcg, beam)

        inner = (output.conj() * target_field).sum().abs()
        norm_out = output.abs().square().sum().sqrt()
        overlap_t = inner / (norm_out * norm_tgt).clamp(min=1e-30)
        overlap_floor = torch.tensor(V92_STAGE_C_OVERLAP_RETENTION, device=device, dtype=torch.float64)
        overlap_loss = torch.relu(overlap_floor - overlap_t).square()
        phase_soft = torch.relu(ph_var - soft_var)
        phase_hard = torch.relu(ph_var - hard_var)

        loss = (
            V92_STAGE_C_UNIFORMITY_WEIGHT * uni_loss
            + V92_STAGE_C_COST_WEIGHT * (cost_se / init_cost_scale)
            + V92_STAGE_C_SPILL_WEIGHT * spill
            + V92_STAGE_C_PHASE_SOFT_WEIGHT * phase_soft
            + V92_STAGE_C_PHASE_HARD_WEIGHT * phase_hard
            + V92_STAGE_C_OVERLAP_WEIGHT * overlap_loss
        )
        loss.backward()
        torch.nn.utils.clip_grad_norm_([phase], V92_STAGE_C_GRAD_CLIP)
        optimizer.step()
        with torch.no_grad():
            phase.copy_(torch.remainder(phase + math.pi, 2 * math.pi) - math.pi)

        if step == 1 or step % V92_STAGE_C_EVAL_EVERY == 0 or step == V92_STAGE_C_STEPS:
            with torch.no_grad():
                out_eval = propagate(beam * torch.exp(1j * phase))
                overlap = compute_overlap(out_eval, target_field)
                cost_se_val = ((Ta_sq - out_eval.abs().square()) * plateau_mask).square().sum().item()
                phase_flat = phase_flatness_metric(out_eval, target_amp, frac=0.5)
                flat_rms_val, _, _ = proper_uniformity(out_eval.abs(), target_amp, frac=0.9)
                overlap_retention = overlap / max(init_overlap, 1e-30)
                flat_finite = _v92_value_is_finite(flat_rms_val)
                phase_ok = phase_flat <= V92_STAGE_C_HARD_PHASE
                overlap_ok = overlap_retention >= V92_STAGE_C_OVERLAP_RETENTION
                flat_ok = flat_finite and flat_rms_val <= best_flat_rms * V92_STAGE_C_FLAT_IMPROVE
                print(f"  [v92-C] step {step:02d}/{V92_STAGE_C_STEPS}: loss={loss.item():.3e}, "
                      f"cost_se={cost_se_val:.2e}, overlap={overlap:.4f}, phase={phase_flat:.3f}, "
                      f"flat_rms={flat_rms_val:.3f}, overlap_ret={overlap_retention*100:.1f}%")
                if phase_ok and overlap_ok and flat_ok:
                    best_phase = phase.detach().clone()
                    best_flat_rms = flat_rms_val
                    best_overlap = overlap
                    best_phase_flat = phase_flat
                    best_cost_se = cost_se_val
                    phase_candidates.append((f"C_step{step:02d}", best_phase.clone()))
                    print(f"    Accepted: flat_rms -> {best_flat_rms:.3f}, phase={best_phase_flat:.3f}, overlap={best_overlap:.4f}")
                elif phase_flat > V92_STAGE_C_HARD_PHASE * 1.08:
                    print(f"    Stop: phase {phase_flat:.3f} exceeded hard cap margin")
                    break

    elapsed = time.time() - t0
    print(f"  [v92-C] done: flat_rms {init_flat_rms:.3f} -> {best_flat_rms:.3f}, "
          f"overlap {init_overlap:.4f} -> {best_overlap:.4f}, "
          f"phase {init_phase_flat:.3f} -> {best_phase_flat:.3f}rad, cost_se {init_cost_se:.2e} -> {best_cost_se:.2e}, time={elapsed:.1f}s")
    return best_phase, elapsed, phase_candidates


def _inv_propagate(field_focal):
    """Inverse of propagate(): focal field (OUT_H, OUT_W) -> SLM-plane field (OUT_H, OUT_W)."""
    oh, ow = field_focal.shape
    A0 = 1.0 / math.sqrt(oh * ow)
    slm_full = torch.fft.ifftshift(torch.fft.ifft2(torch.fft.fftshift(field_focal))) / A0
    return slm_full


def run_v92_stage_d(phase_in, beam, target_amp, Wcg, target_field):
    """v92 Stage D: damped WGS with per-round alpha backtracking."""
    print(f"\n  [v92-D] WGS alpha backtracking (alphas={V92_STAGE_D_ALPHAS}, cap={V92_STAGE_D_PHASE_CAP}, rounds={V92_STAGE_D_ROUNDS})")
    h, w = phase_in.shape
    pad_h = (OUT_H - h) // 2
    pad_w = (OUT_W - w) // 2
    Wcg_mask = (Wcg > 0)

    phase = phase_in.detach().clone()
    with torch.no_grad():
        out0 = propagate(beam * torch.exp(1j * phase))
        init_overlap = compute_overlap(out0, target_field)
        init_phase_flat = phase_flatness_metric(out0, target_amp, frac=0.5)
        init_flat_rms, _, _ = proper_uniformity(out0.abs(), target_amp, frac=0.9)
        init_eff = (out0.abs().square()[Wcg > 0].sum() / beam.square().sum()).item()
        init_spill, _ = spillover_ratio(out0.abs(), Wcg, beam)
    print(f"  [v92-D] init: overlap={init_overlap:.4f}, phase={init_phase_flat:.3f}rad, "
          f"flat_rms={init_flat_rms:.3f}, eff={init_eff:.4f}, spill={init_spill:.3f}")

    best_phase = phase.clone()
    best_flat_rms = init_flat_rms
    best_spill = init_spill
    phase_candidates = [("D_init", best_phase.clone())]
    t0 = time.time()

    with torch.no_grad():
        for rd in range(1, V92_STAGE_D_ROUNDS + 1):
            out = propagate(beam * torch.exp(1j * phase))
            out_phase = torch.angle(out)
            current_amp = out.abs()
            valid_candidates = []

            for alpha in V92_STAGE_D_ALPHAS:
                damped_amp = alpha * target_amp + (1.0 - alpha) * current_amp
                target_field_focal = damped_amp.to(out.real.dtype) * torch.exp(1j * out_phase)
                out_new = torch.where(Wcg_mask, target_field_focal, out)
                slm_full = _inv_propagate(out_new)
                slm_crop = slm_full[pad_h:pad_h + h, pad_w:pad_w + w]
                phase_candidate = torch.angle(slm_crop)

                out_check = propagate(beam * torch.exp(1j * phase_candidate))
                ov = compute_overlap(out_check, target_field)
                ph = phase_flatness_metric(out_check, target_amp, frac=0.5)
                rms, _, _ = proper_uniformity(out_check.abs(), target_amp, frac=0.9)
                eff = (out_check.abs().square()[Wcg > 0].sum() / beam.square().sum()).item()
                spill, _ = spillover_ratio(out_check.abs(), Wcg, beam)
                rms_finite = _v92_value_is_finite(rms)
                valid = (
                    rms_finite
                    and ph <= V92_STAGE_D_PHASE_CAP
                    and rms <= best_flat_rms * V92_STAGE_D_MIN_IMPROVEMENT
                    and ov >= init_overlap * V92_STAGE_D_OVERLAP_RETENTION
                    and eff >= init_eff * V92_STAGE_D_EFF_RETENTION
                    and spill <= max(best_spill, 1e-12) * V92_STAGE_D_SPILL_MAX_GROWTH
                )
                print(f"    [D rd{rd} alpha={alpha:.2f}] overlap={ov:.4f}, phase={ph:.3f}, flat_rms={rms:.3f}, eff={eff:.4f}, spill={spill:.3f}, valid={valid}")
                if valid:
                    score = rms + 0.15 * (ph / V92_STAGE_D_PHASE_CAP) + 0.05 * (spill / max(best_spill, 1e-12))
                    valid_candidates.append((score, alpha, phase_candidate.clone(), rms, ph, ov, eff, spill))

            if not valid_candidates:
                print(f"  [v92-D] round {rd}: no alpha passed guard; stop")
                break

            score, alpha, phase, best_flat_rms, ph, ov, eff, best_spill = min(valid_candidates, key=lambda x: x[0])
            best_phase = phase.clone()
            phase_candidates.append((f"D_rd{rd}_a{alpha:.2f}", best_phase.clone()))
            print(f"  [v92-D] round {rd}: accepted alpha={alpha:.2f}, overlap={ov:.4f}, phase={ph:.3f}, flat_rms={best_flat_rms:.3f}, eff={eff:.4f}")

    elapsed = time.time() - t0
    print(f"  [v92-D] done: flat_rms {init_flat_rms:.3f} -> {best_flat_rms:.3f}, time={elapsed:.1f}s")
    return best_phase, elapsed, phase_candidates


def run_pipeline_v92(beam, target_amp, Wcg, target_field, norm_tgt, target_phase, priority_w):
    """v92 Pipeline: Stage B snapshot pool -> Adam Stage C -> alpha-backtracking Stage D -> Pareto representatives."""
    print(f"\n--- v92 Pipeline (StageB snapshots -> Adam C trust-region -> WGS D backtracking) ---")
    print(f"\n  [v92-init] LG init seed=42 (skip Stage A)")
    lg_init = make_lg_phase(42)
    candidate_phases = [("B_LG_init", lg_init.detach().clone())]
    with torch.no_grad():
        out0 = propagate(beam * torch.exp(1j * lg_init))
        ov0 = compute_overlap(out0, target_field)
        ph0 = phase_flatness_metric(out0, target_amp, frac=0.5)
    print(f"  [v92-init] LG: overlap={ov0:.4f}, phase_flat={ph0:.3f}rad")

    print(f"\n  [v92-B] fmin_cg on (1-overlap)^2 * (1/eff)^{V92_STAGE_B_EFF_POWER}, {V92_STAGE_B_ITERS} iter; tracking multiple snapshots")
    phase_np = lg_init.detach().cpu().numpy().flatten().copy()
    total_input = beam.square().sum()
    eval_count = [0]
    b_tracks = {
        "B_cost_min": {"phase": None, "cost": float("inf")},
        "B_phase_min": {"phase": None, "phase_flat": float("inf")},
        "B_overlap_max": {"phase": None, "overlap": -float("inf")},
        "B_eff_max": {"phase": None, "eff": -float("inf")},
        "B_constrained_best": {"phase": None, "cost": float("inf"), "phase_flat": float("inf")},
    }

    def _save_b_track(name, phi_flat, **values):
        b_tracks[name]["phase"] = phi_flat.copy()
        b_tracks[name].update(values)

    def cost_and_grad_b(phi_flat):
        phase_t = torch.from_numpy(phi_flat).to(device=device, dtype=torch.float64).reshape(SLM_H, SLM_W)
        phase_t = phase_t.requires_grad_(True)
        output = propagate(beam * torch.exp(1j * phase_t))
        E_out_2 = output.abs().square()
        inner = (output.conj() * target_field).sum().abs()
        norm_out = output.abs().square().sum().sqrt()
        overlap = inner / (norm_out * norm_tgt).clamp(min=1e-30)
        eff = (E_out_2[Wcg > 0].sum() / total_input.clamp(min=1e-30)).clamp(min=1e-6)
        cost = 1e9 * (1.0 - overlap).square() * torch.pow(1.0 / eff, V92_STAGE_B_EFF_POWER)
        cost.backward()
        grad = phase_t.grad.detach().cpu().numpy().flatten()
        eval_count[0] += 1
        cv = cost.item()
        ov = overlap.item()
        ef = eff.item()
        with torch.no_grad():
            ph = phase_flatness_metric(output, target_amp, frac=0.5)
        if cv < b_tracks["B_cost_min"]["cost"]:
            _save_b_track("B_cost_min", phi_flat, cost=cv, phase_flat=ph, overlap=ov, eff=ef)
        if ph < b_tracks["B_phase_min"]["phase_flat"]:
            _save_b_track("B_phase_min", phi_flat, cost=cv, phase_flat=ph, overlap=ov, eff=ef)
        if ov > b_tracks["B_overlap_max"]["overlap"]:
            _save_b_track("B_overlap_max", phi_flat, cost=cv, phase_flat=ph, overlap=ov, eff=ef)
        if ef > b_tracks["B_eff_max"]["eff"]:
            _save_b_track("B_eff_max", phi_flat, cost=cv, phase_flat=ph, overlap=ov, eff=ef)
        if ph <= V92_PHASE_THRESHOLD_B and cv < b_tracks["B_constrained_best"]["cost"]:
            _save_b_track("B_constrained_best", phi_flat, cost=cv, phase_flat=ph, overlap=ov, eff=ef)
        if eval_count[0] % 20 == 0:
            mark = "*" if ph <= V92_PHASE_THRESHOLD_B else " "
            print(f"    [B] eval {eval_count[0]}{mark}: cost={cv:.2e}, overlap={ov:.4f}, eff={ef:.4f}, phase_flat={ph:.3f}rad")
        return cv, grad

    cost_fn, grad_fn = _v89_make_cached_cost_grad(cost_and_grad_b)
    t0 = time.time()
    try:
        scipy.optimize.fmin_cg(f=cost_fn, x0=phase_np, fprime=grad_fn,
                               maxiter=V92_STAGE_B_ITERS, disp=False)
    except Exception as e:
        print(f"    [B] fmin_cg early stop: {e}")
    time_b = time.time() - t0

    for name, rec in b_tracks.items():
        if rec["phase"] is not None:
            phase_track = torch.from_numpy(rec["phase"]).to(device=device, dtype=torch.float64).reshape(SLM_H, SLM_W)
            candidate_phases.append((name, phase_track))
            print(f"    [{name}] overlap={rec.get('overlap', float('nan')):.4f}, eff={rec.get('eff', float('nan')):.4f}, phase={rec.get('phase_flat', float('nan')):.3f}, cost={rec.get('cost', float('nan')):.2e}")

    selected_b = b_tracks["B_constrained_best"] if b_tracks["B_constrained_best"]["phase"] is not None else b_tracks["B_cost_min"]
    phase_b = torch.from_numpy(selected_b["phase"]).to(device=device, dtype=torch.float64).reshape(SLM_H, SLM_W)
    with torch.no_grad():
        out = propagate(beam * torch.exp(1j * phase_b))
        ov = compute_overlap(out, target_field)
        ph = phase_flatness_metric(out, target_amp, frac=0.5)
        eff = (out.abs().square()[Wcg > 0].sum() / beam.square().sum()).item()
        rms, _, _ = proper_uniformity(out.abs(), target_amp, frac=0.9)
    print(f"  [v92-B] selected entry: overlap={ov:.4f}, phase={ph:.3f}rad, eff={eff:.4f}, flat_rms={rms:.3f}, evals={eval_count[0]}, time={time_b:.1f}s")

    phase_c, time_c, c_candidates = run_v92_stage_c(phase_b, beam, target_amp, Wcg, target_field, norm_tgt)
    candidate_phases.extend(c_candidates)

    phase_d, time_d, d_candidates = run_v92_stage_d(phase_c, beam, target_amp, Wcg, target_field)
    candidate_phases.extend(d_candidates)

    candidates = []
    seen = set()
    for name, ph_tensor in candidate_phases:
        if name in seen:
            continue
        seen.add(name)
        metrics = _v92_candidate_metrics(ph_tensor, beam, target_amp, target_phase, Wcg, priority_w)
        candidates.append({"name": name, "phase": ph_tensor.detach().clone(), "metrics": metrics})
    _v92_print_candidate_table(candidates)
    reps = _v92_select_representatives(candidates)
    _v92_print_representatives(reps)
    selected = reps["Pipeline_balanced"]

    total = time_b + time_c + time_d
    print(f"\n  [v92] selected Pipeline_balanced = {selected['name']}; total time B={time_b:.1f}s + C={time_c:.1f}s + D={time_d:.1f}s = {total:.1f}s")
    return selected["phase"], total




def _v93_tv_grad(phase):
    """Approximate gradient of total-variation on a 2D phase map (L1 TV)."""
    dy = phase[1:, :] - phase[:-1, :]
    dx = phase[:, 1:] - phase[:, :-1]
    sx = torch.sign(dx)
    sy = torch.sign(dy)
    grad = torch.zeros_like(phase)
    # gradient contributions from |dy|: each term affects two rows
    grad[1:, :] += sy
    grad[:-1, :] -= sy
    grad[:, 1:] += sx
    grad[:, :-1] -= sx
    return grad


def _v93_circ_diff(a, b):
    """Wrapped phase difference in (-pi, pi]."""
    d = a - b
    return torch.atan2(torch.sin(d), torch.cos(d))


def run_v93_stage_0(beam, target_amp, Wcg, target_field, norm_tgt):
    """v93 Stage 0: Bowman warm-start.
    LG init + scipy fmin_cg on cost_se, track-by-overlap snapshot.
    Returns phase_b0 with phase_flat ~0.28 rad.
    """
    print(f"\n  [v93-0] Bowman warm-start: LG seed={V93_STAGE0_LG_SEED} + fmin_cg(cost_se), "
          f"{V93_STAGE0_ITERS} maxiter, snapshot best overlap")
    Ta_sq = target_amp.square()
    init_phase = make_lg_phase(V93_STAGE0_LG_SEED)
    phase_np = init_phase.detach().cpu().numpy().flatten().copy()
    eval_count = [0]
    best_overlap = [-1.0]
    best_phase_arr = [phase_np.copy()]
    best_phase_flat = [float("inf")]

    def cost_and_grad(phi_flat):
        phase_t = torch.from_numpy(phi_flat).to(device=device, dtype=torch.float64).reshape(SLM_H, SLM_W)
        phase_t = phase_t.requires_grad_(True)
        output = propagate(beam * torch.exp(1j * phase_t))
        E_out_2 = output.abs().square()
        cost = ((Ta_sq - E_out_2) * Wcg).square().sum()
        cost.backward()
        grad = phase_t.grad.detach().cpu().numpy().flatten()
        eval_count[0] += 1
        with torch.no_grad():
            ov = compute_overlap(output, target_field)
            ph = phase_flatness_metric(output, target_amp, frac=0.5)
            if ov > best_overlap[0]:
                best_overlap[0] = ov
                best_phase_arr[0] = phi_flat.copy()
                best_phase_flat[0] = ph
        if eval_count[0] % 30 == 0:
            print(f"    [0] eval {eval_count[0]}: cost={cost.item():.2e}, overlap={ov:.4f}, phase_flat={ph:.3f}rad")
        return cost.item(), grad

    cost_fn, grad_fn = _v89_make_cached_cost_grad(cost_and_grad)
    t0 = time.time()
    try:
        scipy.optimize.fmin_cg(f=cost_fn, x0=phase_np, fprime=grad_fn,
                               maxiter=V93_STAGE0_ITERS, disp=False)
    except Exception as e:
        print(f"    [0] fmin_cg early stop: {e}")
    elapsed = time.time() - t0
    phase_b0 = torch.from_numpy(best_phase_arr[0]).to(device=device, dtype=torch.float64).reshape(SLM_H, SLM_W)
    with torch.no_grad():
        out = propagate(beam * torch.exp(1j * phase_b0))
        ov = compute_overlap(out, target_field)
        ph = phase_flatness_metric(out, target_amp, frac=0.5)
        eff = (out.abs().square()[Wcg > 0].sum() / beam.square().sum()).item()
    print(f"  [v93-0] 完成: best_overlap={ov:.4f}, phase_flat={ph:.3f}rad, eff={eff:.4f}, "
          f"evals={eval_count[0]}, 耗时={elapsed:.1f}s")
    return phase_b0, elapsed


def run_v93_stage_a(phase_b0, beam, target_amp, Wcg, target_field, norm_tgt):
    """v93 Stage A: anchored efficiency lift.
    Loss = 1e9·(1-overlap)²·(1/eff)^k + λ_anchor·mean(wrap(phase - phase_b0)²)
    λ_anchor decays geometrically.
    """
    print(f"\n  [v93-A] anchored eff lift: fmin_cg, {V93_STAGEA_ITERS} maxiter, "
          f"k={V93_STAGEA_EFF_POWER}, λ_init={V93_STAGEA_LAMBDA_INIT}, decay every {V93_STAGEA_LAMBDA_DECAY_EVERY} evals")
    phase_np = phase_b0.detach().cpu().numpy().flatten().copy()
    phase_b0_const = phase_b0.detach().clone()
    total_input = beam.square().sum()
    eval_count = [0]
    lam_state = [V93_STAGEA_LAMBDA_INIT]
    best = {"phase": phase_np.copy(), "cost": float("inf"), "phase_flat": float("inf"), "overlap": -1.0, "eff": 0.0}

    def cost_and_grad(phi_flat):
        # Decay anchor weight
        if eval_count[0] > 0 and eval_count[0] % V93_STAGEA_LAMBDA_DECAY_EVERY == 0:
            lam_state[0] = max(V93_STAGEA_LAMBDA_MIN, lam_state[0] * 0.5)
        phase_t = torch.from_numpy(phi_flat).to(device=device, dtype=torch.float64).reshape(SLM_H, SLM_W)
        phase_t = phase_t.requires_grad_(True)
        output = propagate(beam * torch.exp(1j * phase_t))
        E_out_2 = output.abs().square()
        inner = (output.conj() * target_field).sum().abs()
        norm_out = output.abs().square().sum().sqrt()
        overlap = inner / (norm_out * norm_tgt).clamp(min=1e-30)
        eff = (E_out_2[Wcg > 0].sum() / total_input.clamp(min=1e-30)).clamp(min=1e-6)
        # anchor distance (wrapped)
        dphi = _v93_circ_diff(phase_t, phase_b0_const)
        anchor_loss = dphi.square().mean()
        cost = (1e9 * (1.0 - overlap).square() * torch.pow(1.0 / eff, V93_STAGEA_EFF_POWER)
                + lam_state[0] * 1e7 * anchor_loss)
        cost.backward()
        grad = phase_t.grad.detach().cpu().numpy().flatten()
        eval_count[0] += 1
        cv = cost.item()
        # v95 P1-3: numerical-stability guard — if cost explodes (line search misfire),
        # return huge cost + zero grad so fmin_cg backs off this region.
        import math as _math
        if not _math.isfinite(cv) or cv > 1.0e10 or _math.isnan(cv):
            print(f"    [A] eval {eval_count[0]} ✗ numerical instability (cv={cv:.2e}); returning inf")
            return 1.0e15, np.zeros_like(grad)
        with torch.no_grad():
            ph = phase_flatness_metric(output, target_amp, frac=0.5)
        # snapshot best by cost (anchor naturally bounds phase)
        if cv < best["cost"]:
            best["cost"] = cv
            best["phase"] = phi_flat.copy()
            best["phase_flat"] = ph
            best["overlap"] = overlap.item()
            best["eff"] = eff.item()
        if eval_count[0] % 15 == 0:
            print(f"    [A] eval {eval_count[0]}: cost={cv:.2e}, overlap={overlap.item():.4f}, "
                  f"eff={eff.item():.4f}, phase={ph:.3f}rad, λ={lam_state[0]:.0f}")
        return cv, grad

    cost_fn, grad_fn = _v89_make_cached_cost_grad(cost_and_grad)
    t0 = time.time()
    try:
        scipy.optimize.fmin_cg(f=cost_fn, x0=phase_np, fprime=grad_fn,
                               maxiter=V93_STAGEA_ITERS, disp=False)
    except Exception as e:
        print(f"    [A] fmin_cg early stop: {e}")
    elapsed = time.time() - t0
    phase_a = torch.from_numpy(best["phase"]).to(device=device, dtype=torch.float64).reshape(SLM_H, SLM_W)
    with torch.no_grad():
        out = propagate(beam * torch.exp(1j * phase_a))
        ov = compute_overlap(out, target_field)
        ph = phase_flatness_metric(out, target_amp, frac=0.5)
        eff = (out.abs().square()[Wcg > 0].sum() / beam.square().sum()).item()
        rms, _, _ = proper_uniformity(out.abs(), target_amp, frac=0.9)
    print(f"  [v93-A] 完成: overlap={ov:.4f}, phase={ph:.3f}rad, eff={eff:.4f}, flat_rms={rms:.3f}, "
          f"evals={eval_count[0]}, 耗时={elapsed:.1f}s")
    return phase_a, elapsed


def run_v93_stage_c(phase_in, beam, target_amp, Wcg, target_field, norm_tgt):
    """v93 Stage C: 同 v92 Adam trust-region 但 cap 0.32/0.40 + TV 正则 + 0.93 retention."""
    print(f"\n  [v93-C] Adam trust-region polish v4 (steps={V92_STAGE_C_STEPS}, lr={V92_STAGE_C_LR}, "
          f"soft={V93_STAGEC_SOFT_PHASE}, hard={V93_STAGEC_HARD_PHASE}, "
          f"retention={V93_STAGEC_OVERLAP_RETENTION}, tv={V93_STAGEC_TV_WEIGHT:.0e})")
    Ta_sq = target_amp.square()
    plateau_mask = (target_amp > V92_STAGE_C_PLATEAU_FRAC * target_amp.max()).to(target_amp.dtype) * (Wcg > 0).to(target_amp.dtype)
    if int(plateau_mask.sum().item()) < 10:
        plateau_mask = (Wcg > 0).to(target_amp.dtype)
    tgt_in = Ta_sq[plateau_mask > 0]
    tgt_norm_const = (tgt_in / tgt_in.mean().clamp(min=1e-30)).detach()

    phase = phase_in.detach().clone().requires_grad_(True)
    optimizer = torch.optim.Adam([phase], lr=V92_STAGE_C_LR)
    soft_var = torch.tensor(V93_STAGEC_SOFT_PHASE ** 2, device=device, dtype=torch.float64)
    hard_var = torch.tensor(V93_STAGEC_HARD_PHASE ** 2, device=device, dtype=torch.float64)

    with torch.no_grad():
        output0 = propagate(beam * torch.exp(1j * phase_in))
        init_overlap = compute_overlap(output0, target_field)
        init_cost_se = ((Ta_sq - output0.abs().square()) * plateau_mask).square().sum().item()
        init_phase_flat = phase_flatness_metric(output0, target_amp, frac=0.5)
        init_flat_rms, _, _ = proper_uniformity(output0.abs(), target_amp, frac=0.9)
    init_cost_scale = max(init_cost_se, 1e-30)

    best_phase = phase_in.detach().clone()
    best_flat_rms = init_flat_rms
    best_overlap = init_overlap
    best_phase_flat = init_phase_flat
    phase_candidates = [("C_init", best_phase.clone())]
    t0 = time.time()
    print(f"  [v93-C] init: overlap={init_overlap:.4f}, phase={init_phase_flat:.3f}rad, flat_rms={init_flat_rms:.3f}")

    for step in range(1, V92_STAGE_C_STEPS + 1):
        optimizer.zero_grad()
        output = propagate(beam * torch.exp(1j * phase))
        E_out_2 = output.abs().square()
        cost_se = ((Ta_sq - E_out_2) * plateau_mask).square().sum()
        out_in = E_out_2[plateau_mask > 0]
        out_norm = out_in / out_in.mean().clamp(min=1e-30)
        uni_loss = (out_norm - tgt_norm_const).square().mean()
        ph_var = compute_phase_flatness_loss(output, target_amp, frac=0.5)
        spill = compute_spillover_loss(output, Wcg, beam)

        inner = (output.conj() * target_field).sum().abs()
        norm_out = output.abs().square().sum().sqrt()
        overlap_t = inner / (norm_out * norm_tgt).clamp(min=1e-30)
        overlap_floor = torch.tensor(V93_STAGEC_OVERLAP_RETENTION, device=device, dtype=torch.float64)
        overlap_loss = torch.relu(overlap_floor - overlap_t).square()
        phase_soft = torch.relu(ph_var - soft_var)
        phase_hard = torch.relu(ph_var - hard_var)
        # SLM-side TV regularization (suppress Adam high-freq noise)
        dy = phase[1:, :] - phase[:-1, :]
        dx = phase[:, 1:] - phase[:, :-1]
        tv_loss = dx.abs().mean() + dy.abs().mean()

        loss = (
            V92_STAGE_C_UNIFORMITY_WEIGHT * uni_loss
            + V92_STAGE_C_COST_WEIGHT * (cost_se / init_cost_scale)
            + V92_STAGE_C_SPILL_WEIGHT * spill
            + V92_STAGE_C_PHASE_SOFT_WEIGHT * phase_soft
            + V92_STAGE_C_PHASE_HARD_WEIGHT * phase_hard
            + V92_STAGE_C_OVERLAP_WEIGHT * overlap_loss
            + V93_STAGEC_TV_WEIGHT * tv_loss
        )
        loss.backward()
        torch.nn.utils.clip_grad_norm_([phase], V92_STAGE_C_GRAD_CLIP)
        optimizer.step()
        with torch.no_grad():
            phase.copy_(torch.remainder(phase + math.pi, 2 * math.pi) - math.pi)

        if step == 1 or step % V92_STAGE_C_EVAL_EVERY == 0 or step == V92_STAGE_C_STEPS:
            with torch.no_grad():
                out_eval = propagate(beam * torch.exp(1j * phase))
                ov = compute_overlap(out_eval, target_field)
                ph = phase_flatness_metric(out_eval, target_amp, frac=0.5)
                rms, _, _ = proper_uniformity(out_eval.abs(), target_amp, frac=0.9)
                # v94: relaxed accept — flat improves OR (phase improves AND flat doesn't blow up)
                phase_ok = ph <= V93_STAGEC_HARD_PHASE * 1.01
                overlap_ok = ov >= init_overlap * V93_STAGEC_OVERLAP_RETENTION
                flat_improved = rms < best_flat_rms * 0.99
                phase_improved_amp_stable = (ph < best_phase_flat - 0.005) and (rms <= best_flat_rms * 1.02)
                accept = phase_ok and overlap_ok and (flat_improved or phase_improved_amp_stable)
                if accept:
                    best_flat_rms = min(best_flat_rms, rms)
                    best_phase = phase.detach().clone()
                    best_overlap = ov
                    best_phase_flat = ph
                    snap_name = f"C_step{step:03d}"
                    phase_candidates.append((snap_name, best_phase.clone()))
                    tag = "flat✓" if flat_improved else "phase✓"
                    print(f"    [v94-C] step {step:3d}: overlap={ov:.4f} {tag}, phase={ph:.3f}rad, flat_rms={rms:.3f} (saved {snap_name})")
                else:
                    print(f"    [v94-C] step {step:3d}: overlap={ov:.4f}, phase={ph:.3f}rad, flat_rms={rms:.3f}")

    elapsed = time.time() - t0
    print(f"  [v93-C] 完成: flat_rms {init_flat_rms:.3f} → {best_flat_rms:.3f}, "
          f"overlap {init_overlap:.4f} → {best_overlap:.4f}, "
          f"phase {init_phase_flat:.3f} → {best_phase_flat:.3f}rad, 耗时={elapsed:.1f}s")
    return best_phase, elapsed, phase_candidates


def run_v93_stage_d(phase_in, beam, target_amp, Wcg, target_field):
    """v93 Stage D: damped WGS + per-round TV-prox + alpha backtracking + cap 0.40."""
    print(f"\n  [v93-D] damped WGS + TV-prox (rounds={V92_STAGE_D_ROUNDS}, "
          f"alphas={V93_STAGED_ALPHAS}, cap={V93_STAGED_PHASE_CAP}rad, tv_lr={V93_STAGED_TV_LR})")
    h, w = phase_in.shape
    pad_h = (OUT_H - h) // 2
    pad_w = (OUT_W - w) // 2
    Wcg_mask = (Wcg > 0)

    phase = phase_in.clone()
    with torch.no_grad():
        out = propagate(beam * torch.exp(1j * phase))
        init_overlap = compute_overlap(out, target_field)
        init_phase_flat = phase_flatness_metric(out, target_amp, frac=0.5)
        init_flat_rms, _, _ = proper_uniformity(out.abs(), target_amp, frac=0.9)
        init_eff = (out.abs().square()[Wcg > 0].sum() / beam.square().sum()).item()
        init_spill, _ = spillover_ratio(out.abs(), Wcg, beam)
    print(f"  [v93-D] init: overlap={init_overlap:.4f}, phase={init_phase_flat:.3f}rad, "
          f"flat_rms={init_flat_rms:.3f}, eff={init_eff:.4f}, spill={init_spill:.3f}")

    best_phase = phase.clone()
    best_flat_rms = init_flat_rms
    best_spill = init_spill
    candidates = [("D_init", best_phase.clone())]
    t0 = time.time()

    with torch.no_grad():
        for rd in range(1, V92_STAGE_D_ROUNDS + 1):
            valid_alphas = []
            for alpha in V93_STAGED_ALPHAS:
                # damped WGS step
                field = beam * torch.exp(1j * phase)
                out_f = propagate(field)
                out_amp = out_f.abs()
                out_phase = torch.angle(out_f)
                target_field_focal = (alpha * target_amp + (1.0 - alpha) * out_amp) * torch.exp(1j * out_phase)
                out_new = torch.where(Wcg_mask, target_field_focal, out_f)
                slm_full = _inv_propagate(out_new)
                slm_crop = slm_full[pad_h:pad_h + h, pad_w:pad_w + w]
                phase_cand = torch.angle(slm_crop)
                # TV-prox: 1 step subgradient descent on TV (Gemini 抑毛刺)
                for _ in range(V93_STAGED_TV_STEPS):
                    phase_cand = phase_cand - V93_STAGED_TV_LR * _v93_tv_grad(phase_cand)
                # check candidate
                out_check = propagate(beam * torch.exp(1j * phase_cand))
                ov = compute_overlap(out_check, target_field)
                ph = phase_flatness_metric(out_check, target_amp, frac=0.5)
                rms, _, _ = proper_uniformity(out_check.abs(), target_amp, frac=0.9)
                eff = (out_check.abs().square()[Wcg > 0].sum() / beam.square().sum()).item()
                spill, _ = spillover_ratio(out_check.abs(), Wcg, beam)
                phase_ok = ph <= V93_STAGED_PHASE_CAP
                flat_ok = rms < best_flat_rms * V92_STAGE_D_MIN_IMPROVEMENT
                overlap_ok = ov >= init_overlap * V92_STAGE_D_OVERLAP_RETENTION
                eff_ok = eff >= init_eff * V92_STAGE_D_EFF_RETENTION
                spill_ok = spill <= best_spill * V92_STAGE_D_SPILL_MAX_GROWTH
                if phase_ok and flat_ok and overlap_ok and eff_ok and spill_ok:
                    valid_alphas.append({"alpha": alpha, "phase": phase_cand.clone(),
                                         "rms": rms, "ph": ph, "ov": ov, "eff": eff, "spill": spill})
            if not valid_alphas:
                print(f"  [v93-D] round {rd:2d}: no alpha satisfies all guards, stop")
                break
            best_a = min(valid_alphas, key=lambda c: c["rms"])
            phase = best_a["phase"]
            best_flat_rms = best_a["rms"]
            best_spill = max(best_spill, best_a["spill"])
            best_phase = phase.clone()
            snap_name = f"D_rd{rd:02d}_a{best_a['alpha']:.2f}"
            candidates.append((snap_name, best_phase.clone()))
            print(f"  [v93-D] round {rd:2d} ✓ α={best_a['alpha']:.2f}: "
                  f"overlap={best_a['ov']:.4f}, phase={best_a['ph']:.3f}rad, "
                  f"flat_rms={best_a['rms']:.3f}, eff={best_a['eff']:.4f}")

    elapsed = time.time() - t0
    print(f"  [v93-D] 完成: flat_rms {init_flat_rms:.3f} → {best_flat_rms:.3f}, 耗时={elapsed:.1f}s")
    return best_phase, elapsed, candidates


def _v93_pareto_front(candidates):
    """True non-dominated sorting on (phase_flatness, flat_rms, cost_se, -overlap, -efficiency).
    Returns indices of Pareto-optimal candidates.
    """
    def key_vec(c):
        m = c["metrics"]
        flat = m["flat_rms"] if not (isinstance(m["flat_rms"], float) and math.isnan(m["flat_rms"])) else 1e9
        return (m["phase_flatness"], flat, m["cost_se"], -m["overlap"], -m["efficiency"])

    n = len(candidates)
    front = []
    for i in range(n):
        ki = key_vec(candidates[i])
        dominated = False
        for j in range(n):
            if i == j: continue
            kj = key_vec(candidates[j])
            if all(kj[d] <= ki[d] for d in range(5)) and any(kj[d] < ki[d] for d in range(5)):
                dominated = True
                break
        if not dominated:
            front.append(i)
    return front


def _v93_select_representatives(candidates):
    """Select 4 representatives from Pareto front + balanced (front 中位)."""
    front_idx = _v93_pareto_front(candidates)
    front = [candidates[i] for i in front_idx]
    print(f"\n  [v93-Pareto] candidates total={len(candidates)}, non-dominated front size={len(front)}")
    if not front:
        front = candidates

    def valid(c, k):
        v = c["metrics"][k]
        return not (isinstance(v, float) and math.isnan(v))

    # v94: 阈值放宽 — balanced overlap 0.70->0.65, phase 0.45->0.42; phase_safe overlap 0.70->0.60
    phase_pool = [c for c in front if c["metrics"]["overlap"] >= 0.60]
    phase_safe = min(phase_pool or front, key=lambda c: c["metrics"]["phase_flatness"])

    flat_pool = [c for c in front if valid(c, "flat_rms") and c["metrics"]["phase_flatness"] <= 0.50]
    flat_safe = min(flat_pool or front,
                    key=lambda c: c["metrics"]["flat_rms"] if valid(c, "flat_rms") else 1e9)

    overlap_pool = [c for c in front if c["metrics"]["phase_flatness"] <= 0.50]
    overlap_max = max(overlap_pool or front, key=lambda c: c["metrics"]["overlap"])

    # balanced: front 上加权最优（弱权重 tie-break）
    def balanced_score(c):
        m = c["metrics"]
        flat = m["flat_rms"] if valid(c, "flat_rms") else 3.0
        return (
            min(flat / 0.40, 4.0)
            + min(m["phase_flatness"] / 0.35, 4.0)
            + max(0.0, 0.74 - m["overlap"]) / 0.08
            + max(0.0, 0.55 - m["efficiency"]) / 0.20  # v95: floor 0.70→0.55 (Bowman level), coeff 1/0.10→1/0.20
        )
    balanced_pool = [c for c in front if c["metrics"]["overlap"] >= 0.65 and c["metrics"]["phase_flatness"] <= 0.42]
    balanced = min(balanced_pool or front, key=balanced_score)

    # v94 P1-5: pareto knee — front 上 (phase, flat_rms, cost_se) 归一化距原点最近者
    def _knee_dist(c, front_list):
        m = c["metrics"]
        ph_vals = [x["metrics"]["phase_flatness"] for x in front_list]
        fl_vals = [x["metrics"]["flat_rms"] if valid(x, "flat_rms") else 3.0 for x in front_list]
        cs_vals = [x["metrics"]["cost_se"] for x in front_list]
        ph_min, ph_max = min(ph_vals), max(ph_vals)
        fl_min, fl_max = min(fl_vals), max(fl_vals)
        cs_min, cs_max = min(cs_vals), max(cs_vals)
        n_ph = (m["phase_flatness"] - ph_min) / max(ph_max - ph_min, 1e-9)
        flat = m["flat_rms"] if valid(c, "flat_rms") else 3.0
        n_fl = (flat - fl_min) / max(fl_max - fl_min, 1e-9)
        n_cs = (m["cost_se"] - cs_min) / max(cs_max - cs_min, 1e-9)
        return (n_ph ** 2 + n_fl ** 2 + n_cs ** 2) ** 0.5
    pareto_knee = min(front, key=lambda c: _knee_dist(c, front)) if len(front) > 1 else front[0]

    return {
        "Pipeline_phase_safe": phase_safe,
        "Pipeline_balanced": balanced,
        "Pipeline_flat_safe": flat_safe,
        "Pipeline_overlap_max": overlap_max,
        "Pipeline_pareto_knee": pareto_knee,
        "_pareto_front_idx": front_idx,
    }


def _v93_print_representatives(reps):
    print("\n  [v94-Pareto] representatives (Pipeline_pareto_knee 是真拐点):")
    for cat in ["Pipeline_phase_safe", "Pipeline_balanced", "Pipeline_flat_safe", "Pipeline_overlap_max", "Pipeline_pareto_knee"]:
        c = reps[cat]
        m = c["metrics"]
        flat_str = f"{m['flat_rms']:.3f}" if not (isinstance(m['flat_rms'], float) and math.isnan(m['flat_rms'])) else "nan"
        print(f"    {cat:24s} <- {c['name']:24s} | overlap={m['overlap']:.4f} eff={m['efficiency']:.4f} "
              f"phase={m['phase_flatness']:.3f}rad flat_rms={flat_str} cost_se={m['cost_se']:.0f}")


def run_pipeline_v93(beam, target_amp, Wcg, target_field, norm_tgt, target_phase, priority_w):
    """v93 Pipeline: Stage 0 (Bowman warm) -> Stage A (anchor lift) -> Stage B (multi-snap) -> C v4 -> D v4 -> 真 Pareto."""
    print(f"\n--- v93 Pipeline (Bowman warm-start anchor + 双 cap C + TV-WGS D + 真 Pareto) ---")

    # === Stage 0: Bowman warm-start ===
    phase_b0, time_0 = run_v93_stage_0(beam, target_amp, Wcg, target_field, norm_tgt)
    candidate_phases = [("Stage0_BowmanWarm", phase_b0.detach().clone())]

    # === Stage A: anchored eff lift ===
    phase_a, time_a = run_v93_stage_a(phase_b0, beam, target_amp, Wcg, target_field, norm_tgt)
    candidate_phases.append(("StageA_AnchorLift", phase_a.detach().clone()))

    # === Stage B: 多 snapshot (复用 v92 的 cost_and_grad_b 结构，从 phase_a 起跑) ===
    print(f"\n  [v93-B] fmin_cg on (1-overlap)^2 * (1/eff)^{V92_STAGE_B_EFF_POWER}, {V92_STAGE_B_ITERS} iter, multi-snapshot (start from Stage A output)")
    phase_np = phase_a.detach().cpu().numpy().flatten().copy()
    total_input = beam.square().sum()
    eval_count = [0]
    b_tracks = {
        "B_cost_min": {"phase": None, "cost": float("inf")},
        "B_phase_min": {"phase": None, "phase_flat": float("inf")},
        "B_overlap_max": {"phase": None, "overlap": -float("inf")},
        "B_constrained_best": {"phase": None, "cost": float("inf")},
    }

    def _save_b(name, phi_flat, **values):
        b_tracks[name]["phase"] = phi_flat.copy()
        b_tracks[name].update(values)

    def cost_and_grad_b(phi_flat):
        phase_t = torch.from_numpy(phi_flat).to(device=device, dtype=torch.float64).reshape(SLM_H, SLM_W)
        phase_t = phase_t.requires_grad_(True)
        output = propagate(beam * torch.exp(1j * phase_t))
        E_out_2 = output.abs().square()
        inner = (output.conj() * target_field).sum().abs()
        norm_out = output.abs().square().sum().sqrt()
        overlap = inner / (norm_out * norm_tgt).clamp(min=1e-30)
        eff = (E_out_2[Wcg > 0].sum() / total_input.clamp(min=1e-30)).clamp(min=1e-6)
        cost = 1e9 * (1.0 - overlap).square() * torch.pow(1.0 / eff, V92_STAGE_B_EFF_POWER)
        cost.backward()
        grad = phase_t.grad.detach().cpu().numpy().flatten()
        eval_count[0] += 1
        cv = cost.item()
        ov = overlap.item()
        ef = eff.item()
        with torch.no_grad():
            ph = phase_flatness_metric(output, target_amp, frac=0.5)
        if cv < b_tracks["B_cost_min"]["cost"]:
            _save_b("B_cost_min", phi_flat, cost=cv, phase_flat=ph, overlap=ov, eff=ef)
        if ph < b_tracks["B_phase_min"]["phase_flat"]:
            _save_b("B_phase_min", phi_flat, cost=cv, phase_flat=ph, overlap=ov, eff=ef)
        if ov > b_tracks["B_overlap_max"]["overlap"]:
            _save_b("B_overlap_max", phi_flat, cost=cv, phase_flat=ph, overlap=ov, eff=ef)
        if ph <= V92_PHASE_THRESHOLD_B and cv < b_tracks["B_constrained_best"]["cost"]:
            _save_b("B_constrained_best", phi_flat, cost=cv, phase_flat=ph, overlap=ov, eff=ef)
        if eval_count[0] % 20 == 0:
            mark = "*" if ph <= V92_PHASE_THRESHOLD_B else " "
            print(f"    [B] eval {eval_count[0]}{mark}: cost={cv:.2e}, overlap={ov:.4f}, eff={ef:.4f}, phase={ph:.3f}rad")
        return cv, grad

    cost_fn, grad_fn = _v89_make_cached_cost_grad(cost_and_grad_b)
    t0 = time.time()
    try:
        scipy.optimize.fmin_cg(f=cost_fn, x0=phase_np, fprime=grad_fn, maxiter=V92_STAGE_B_ITERS, disp=False)
    except Exception as e:
        print(f"    [B] fmin_cg early stop: {e}")
    time_b = time.time() - t0
    for name, rec in b_tracks.items():
        if rec["phase"] is not None:
            ph_track = torch.from_numpy(rec["phase"]).to(device=device, dtype=torch.float64).reshape(SLM_H, SLM_W)
            candidate_phases.append((name, ph_track))

    selected_b = b_tracks["B_constrained_best"] if b_tracks["B_constrained_best"]["phase"] is not None else b_tracks["B_cost_min"]
    phase_b = torch.from_numpy(selected_b["phase"]).to(device=device, dtype=torch.float64).reshape(SLM_H, SLM_W)
    print(f"  [v93-B] selected entry: phase={selected_b.get('phase_flat', float('nan')):.3f}rad, "
          f"overlap={selected_b.get('overlap', float('nan')):.4f}, evals={eval_count[0]}, time={time_b:.1f}s")

    # === Stage C v4 ===
    phase_c, time_c, c_cands = run_v93_stage_c(phase_b, beam, target_amp, Wcg, target_field, norm_tgt)
    candidate_phases.extend(c_cands)

    # === Stage D v4 ===
    phase_d, time_d, d_cands = run_v93_stage_d(phase_c, beam, target_amp, Wcg, target_field)
    candidate_phases.extend(d_cands)

    # === True non-dominated Pareto ===
    candidates = []
    seen = set()
    for name, ph_tensor in candidate_phases:
        if name in seen:
            continue
        seen.add(name)
        metrics = _v92_candidate_metrics(ph_tensor, beam, target_amp, target_phase, Wcg, priority_w)
        candidates.append({"name": name, "phase": ph_tensor.detach().clone(), "metrics": metrics})
    _v92_print_candidate_table(candidates)
    reps = _v93_select_representatives(candidates)
    _v93_print_representatives(reps)
    selected = reps["Pipeline_pareto_knee"]  # v95: switch from balanced to true Pareto knee

    total = time_0 + time_a + time_b + time_c + time_d
    print(f"\n  [v93] selected = {selected['name']}; total time 0={time_0:.1f}s + A={time_a:.1f}s + "
          f"B={time_b:.1f}s + C={time_c:.1f}s + D={time_d:.1f}s = {total:.1f}s")
    return selected["phase"], total


def run_pipeline(beam, target_amp, Wcg, target_field, norm_tgt):
    print(f"\n--- Pipeline (两阶段效率优先, 相位约束, 5维 Rank-Pareto) ---")
    print(f"  阶段1: {STAGE1_ITERS} iter, alpha={STAGE1_ALPHA}, lr={STAGE1_LR} (overlap优先)")
    print(f"阶段1: spill_s1={SPILLOVER_ALPHA_S1}, 阶段2: alpha={STAGE2_ALPHA}, lr={STAGE2_LR}, ph={PHASE_FLATNESS_ALPHA}, spill_s2={SPILLOVER_ALPHA_S2} (效率+相位约束)")

    curv_bases = [1.0, 2.0, 3.0, 5.0]
    charge_spreads = [0.1, 0.2, 0.1, 0.3]
    candidate_seeds = list(range(100, 100 + NUM_SEEDS_CANDIDATE))

    quick_results = []
    total_quick_time = 0
    for idx, seed in enumerate(candidate_seeds):
        cb = curv_bases[idx % len(curv_bases)]
        cs = charge_spreads[idx % len(charge_spreads)]
        phase_init = make_lg_phase(seed, curv_base=cb, charge_spread=cs)
        phase, _, quick_time = run_adam_single_stage(
            phase_init, beam, target_amp, Wcg, target_field, norm_tgt,
            num_iters=SEED_QUICK_ITERS, lr=STAGE1_LR, alpha=STAGE1_ALPHA,
            label=f"Quick s{seed}", use_priority=False, use_efficiency=False)
        total_quick_time += quick_time
        with torch.no_grad():
            field = beam * torch.exp(1j * phase)
            output = propagate(field)
            overlap = compute_overlap(output, target_field)
        quick_results.append((seed, cb, cs, phase, overlap))
        print(f"  [Quick] seed={seed} curv={cb} charge_spread={cs} -> overlap={overlap:.6f}")

    quick_results.sort(key=lambda x: x[4], reverse=True)
    selected = quick_results[:NUM_SEEDS_FINAL]
    print(f"\n  Selected seeds: {[s[0] for s in selected]}")

    all_candidates = []
    total_adam_time = 0
    for seed, cb, cs, quick_phase, _ in selected:
        phase_init = make_lg_phase(seed, curv_base=cb, charge_spread=cs)
        ps1, o1, t1, ps2, o2, t2 = run_two_stage_adam(
            phase_init, beam, target_amp, Wcg, target_field, norm_tgt,
            label=f"s{seed}")
        total_adam_time += t1 + t2
        all_candidates.append((f"s{seed}_stage1", ps1, o1, t1))
        all_candidates.append((f"s{seed}_stage2", ps2, o2, t2))
        
    print(f"\n  Pipeline Adam 完成, quick={total_quick_time:.1f}s, full={total_adam_time:.1f}s")

    # v82: 5-dim Rank-Pareto over Stage 2+3 candidates (all optimize phase)
    print(f"\n  Stage 2+3 5维 Rank-Pareto 选择 (overlap+eff+cost_se+phase_flatness+flat_rms):")
    priority_w = make_priority_weight(target_amp, Wcg)
    # v84: include stage1 candidates — Stage 1 often has better phase (0.45 rad)
    # and Pareto weight phase*3 will prefer it over stage2 (0.58 rad)
    pareto_candidates = all_candidates
    best_name, best_phase, best_metrics, best_time, _ = select_pareto_candidate(
        pareto_candidates, beam, target_amp, target_phase, Wcg, priority_w)

    print(f"\n  L-BFGS polish ({best_name})...")
    best_phase, lbfgs_time = run_lbfgs_polish(
        best_phase, beam, target_amp, Wcg, target_field, label="Pipe L-BFGS")

    print(f"\n  Cost-SE L-BFGS 精修 (含 phase loss + 收紧接受准则)...")
    best_phase, cost_lbfgs_time = run_cost_se_lbfgs(
        best_phase, beam, target_amp, Wcg, target_field, label="Pipe CostSE-LBFGS")

    total_time = total_quick_time + total_adam_time + lbfgs_time + cost_lbfgs_time
    return best_phase, total_time


print("=" * 60)
print(f"v95 three-algorithm comparison (= v94 algo, switch selected to Pareto knee + eff weight relaxed)")
print("=" * 60)

beam = make_beam()
target_amp, target_phase, Wcg = make_target_and_weight()
target_field = target_amp * torch.exp(1j * target_phase)
norm_tgt = target_field.abs().square().sum().sqrt().detach()
priority_w = make_priority_weight(target_amp, Wcg)
print(f"\n光束形状: {beam.shape}, 目标形状: {target_amp.shape}")
print(f"光束能量: {beam.square().sum().item():.6f}")
print(f"Wcg 非零像素: {(Wcg > 0).sum().item()}")

phase_795, time_795 = run_795_cg(beam, target_amp, Wcg, target_field, norm_tgt)
metrics_795 = evaluate_metrics(phase_795, beam, target_amp, target_phase, Wcg, priority_w)

phase_bowman, time_bowman = run_bowman_cg(beam, target_amp, Wcg, target_field, norm_tgt)
metrics_bowman = evaluate_metrics(phase_bowman, beam, target_amp, target_phase, Wcg, priority_w)

phase_pipe, time_pipe = run_pipeline_v93(beam, target_amp, Wcg, target_field, norm_tgt, target_phase, priority_w)
metrics_pipe = evaluate_metrics(phase_pipe, beam, target_amp, target_phase, Wcg, priority_w)

print(f"\n" + "=" * 100)
print(f"v95 三算法对比结果 (Pipeline = v94 algo + selected: Pareto knee)")
print("=" * 100)

all_results = [
    ("795 CG", metrics_795, time_795),
    ("Bowman CG", metrics_bowman, time_bowman),
    ("Pipeline", metrics_pipe, time_pipe),
]

metric_keys = ["overlap", "efficiency", "wcg_corr",
               "flat_rms", "flat_pv", "phase_flatness", "spillover", "in_band",
               "core_uniformity", "cost_se"]
metric_labels = metric_keys

header = f"{'指标':<20}" + "".join(f"{name:>15}" for name, _, _ in all_results)
print(header)
print("-" * len(header))

for key, label in zip(metric_keys, metric_labels):
    row = f"{label:<20}"
    for name, m, _ in all_results:
        v = m[key]
        if isinstance(v, float) and math.isnan(v):
            row += f"{'nan':>15}"
        else:
            row += f"{v:>15.6f}"
    print(row)

row = f"{'耗时(s)':<20}"
for name, _, t in all_results:
    row += f"{t:>15.1f}"
print(row)


def _safe_min(results, key):
    cands = [r for r in results if not (isinstance(r[1][key], float) and math.isnan(r[1][key]))]
    if not cands:
        return "n/a"
    return min(cands, key=lambda x: x[1][key])[0]


def _safe_max(results, key):
    cands = [r for r in results if not (isinstance(r[1][key], float) and math.isnan(r[1][key]))]
    if not cands:
        return "n/a"
    return max(cands, key=lambda x: x[1][key])[0]


print(f"\n最优 overlap: {_safe_max(all_results, 'overlap')}")
print(f"最优 efficiency: {_safe_max(all_results, 'efficiency')}")
print(f"最优 cost_se: {_safe_min(all_results, 'cost_se')}")
print(f"最优 flat_rms (平顶 RMS): {_safe_min(all_results, 'flat_rms')}")
print(f"最优 flat_pv (平顶 PV): {_safe_min(all_results, 'flat_pv')}")
print(f"最优 phase_flatness (平顶相位 std rad, 目标<0.3): {_safe_min(all_results, 'phase_flatness')}")
print(f"最优 spillover (圆盘外能量比): {_safe_min(all_results, 'spillover')}")

print(f"\n--- v92 update notes (trust-region + candidate pool) ---")
print(f"1. Stage B keeps the v91 loss but records cost/phase/overlap/eff/constrained snapshots")
print(f"2. Stage C switches from L-BFGS to Adam micro-polish: steps={V92_STAGE_C_STEPS}, lr={V92_STAGE_C_LR}, hard_phase={V92_STAGE_C_HARD_PHASE}")
print(f"3. Stage C loss includes normalized cost, uniformity, spill, soft/hard phase, and overlap retention")
print(f"4. Stage D uses alpha backtracking: alphas={V92_STAGE_D_ALPHAS}, phase_cap={V92_STAGE_D_PHASE_CAP}")
print(f"5. Pipeline prints candidate pool and phase/balanced/flat/cost representatives; final table uses Pipeline_balanced")
print(f"6. Expected target: overlap>=0.72, eff>=0.72, phase<=0.45, flat_rms<=0.40, cost_se<1e4")
````

## 分析 quality 路由结果
基于 notebook 和 `summary.json` 分析结果
1. 按照 `core_phase_flatness -> core_uniformity_loss -> overlap_loss` 排序路由
2. quality 路由使用了哪个 checkpoint?
3. 为什么这个路由被选为 quality 路由?
4. efficiency 路由的表现如何?


````python
from pathlib import Path
import json


def route_score(metrics: dict) -> float:
    return (
        2.0 * float(metrics.get("core_phase_flatness", float("inf")))
        + 0.6 * float(metrics.get("core_uniformity_loss", float("inf")))
        + 0.1 * float(metrics.get("overlap_loss", float("inf")))
    )


def route_rank(metrics: dict) -> tuple[float, float, float, float]:
    return (
        float(metrics.get("core_phase_flatness", float("inf"))),
        float(metrics.get("core_uniformity_loss", float("inf"))),
        float(metrics.get("overlap_loss", float("inf"))),
        route_score(metrics),
    )


def _checkpoint_name(path_value: str | None) -> str:
    return Path(path_value).name if path_value else "default"


def tuning_recommendations(summary: dict) -> dict:
    routes = summary.get("routes", {})
    best_route = summary.get("best_route")
    best_metrics = routes.get(best_route, {})
    baseline_metrics = routes.get("795_only", {})

    observations = []
    suggestions = []

    if best_route:
        rank = route_rank(best_metrics)
        observations.append(
            f"best route is {best_route}: rank=({rank[0]:.6f}, {rank[1]:.6f}, {rank[2]:.6f}), "
            f"diagnostic_score={rank[3]:.6f}."
        )

    if best_route and best_route.endswith("_quality"):
        observations.append(
            f"quality mainline is active with variant={best_metrics.get('quality_variant', 'baseline')} "
            f"and checkpoint={_checkpoint_name(best_metrics.get('selected_checkpoint'))}."
        )

    if baseline_metrics:
        observations.append(
            "vs 795_only baseline: "
            f"delta core_phase_flatness={float(best_metrics.get('core_phase_flatness', float('inf'))) - float(baseline_metrics.get('core_phase_flatness', float('inf'))):+.6f}, "
            f"delta core_uniformity={float(best_metrics.get('core_uniformity_loss', float('inf'))) - float(baseline_metrics.get('core_uniformity_loss', float('inf'))):+.6f}, "
            f"delta overlap={float(best_metrics.get('overlap', 0.0)) - float(baseline_metrics.get('overlap', 0.0)):+.6f}."
        )

    observations.append(
        "efficiency remains a diagnostic metric only in this benchmark and should not override core phase or core uniformity decisions."
    )

    best_core_phase = float(best_metrics.get("core_phase_flatness", float("inf")))
    baseline_core_phase = float(baseline_metrics.get("core_phase_flatness", float("inf")))
    best_core_uniformity = float(best_metrics.get("core_uniformity_loss", float("inf")))
    baseline_core_uniformity = float(baseline_metrics.get("core_uniformity_loss", float("inf")))
    quality_variant = best_metrics.get("quality_variant", "baseline")

    if baseline_metrics and best_core_phase >= baseline_core_phase:
        suggestions.append(
            {
                "priority": 1,
                "action": "increase phase bias inside the quality family",
                "recommended_changes": {
                    "phase_priority_core_phase_weight": [1.9, 2.1, 2.3],
                    "stage3_maxiter": [12, 14, 16],
                    "init_score_core_phase_weight": [2.4, 2.8, 3.2],
                },
                "why": "the winning route no longer beats the 795 baseline on core_phase_flatness, so the next sweep should push phase harder before touching anything else.",
            }
        )

    if baseline_metrics and best_core_phase < baseline_core_phase and best_core_uniformity > baseline_core_uniformity + 0.0025:
        suggestions.append(
            {
                "priority": 2,
                "action": "recover core uniformity inside the current quality variant",
                "recommended_changes": {
                    "core_uniformity_weight": [0.30, 0.34, 0.38],
                    "phase_post_correction_core_uniformity_weight": [0.04, 0.06, 0.08],
                    "polish_uniformity_weight": [0.02, 0.03, 0.04],
                },
                "why": "core phase is already ahead, so the next clean gain should come from reclaiming core uniformity without giving phase back.",
            }
        )

    if best_route and best_route.endswith("_quality") and quality_variant == "baseline":
        suggestions.append(
            {
                "priority": 1,
                "action": "promote the internal quality sweep to the next comparison run",
                "recommended_values": ["baseline_tightphase", "phase_first"],
                "why": "the baseline quality preset is still winning, so the next step should stay close to the current v59 winner candidate and only test the nearest stronger-phase challengers.",
            }
        )
    elif best_route and best_route.endswith("_quality"):
        suggestions.append(
            {
                "priority": 1,
                "action": "refine locally around the current winning quality variant",
                "recommended_changes": {
                    "baseline_tightphase": {
                        "core_phase_weight": [1.18, 1.22, 1.26],
                        "phase_post_correction_core_phase_weight": [1.00, 1.10, 1.20],
                    },
                    "phase_first": {
                        "stage3_maxiter": [10, 12, 14],
                        "polish_phase_priority_phase_weight": [0.28, 0.30, 0.32],
                    },
                },
                "why": "the winning route already sits inside the quality family, so the next sweep should stay local rather than switching to balanced or feedback_proxy.",
            }
        )

    if not suggestions:
        suggestions.append(
            {
                "priority": 3,
                "action": "continue quality-local sweeps",
                "why": "the current winner already matches the phase-first objective, and no stronger single-metric regression is visible right now.",
            }
        )

    suggestions.sort(key=lambda x: int(x.get("priority", 99)))
    return {"observations": observations, "suggestions": suggestions}


summary_path = Path("ai_holography") / "runs" / "benchmark_lin2025_hybrid" / "summary.json"
summary = json.loads(summary_path.read_text(encoding="utf-8"))
analysis = tuning_recommendations(summary)
summary["autodl_post_analysis"] = analysis
summary_path.write_text(json.dumps(summary, ensure_ascii=False, indent=2), encoding="utf-8")

print(json.dumps(analysis, ensure_ascii=False, indent=2))
````

## 效率上限探测
在当前硬件+几何配置下，只优化 efficiency + overlap（不管相位平坦度），看 efficiency 的物理天花板是多少。
运行后把 `efficiency_floor` 设成 `0.85 × 上限` 即可。


````python
import torch
import torch.nn.functional as F
from ai_holography.config import HolographyConfig
from ai_holography.propagation import FourierSLM, wrap_phase
from ai_holography.targets import build_target, build_weight, threshold_weight, gaussian_beam
from ai_holography.losses import efficiency_metric, normalized_overlap

cfg_eff = HolographyConfig()
cfg_eff.device = "cuda" if torch.cuda.is_available() else "cpu"
device = torch.device(cfg_eff.device)

propagator = FourierSLM(cfg_eff.slm_size, cfg_eff.output_size).to(device)
beam = gaussian_beam(cfg_eff.slm_size, cfg_eff.beam_sigma_x, cfg_eff.beam_sigma_y, cfg_eff.device)
center = (cfg_eff.output_size / 2.0, cfg_eff.output_size / 2.0)
target_amp = build_target("target_flat_top", cfg_eff.output_size, center, cfg_eff.target_sigma, 1, cfg_eff.device)
target_phase = torch.zeros((cfg_eff.output_size, cfg_eff.output_size), device=device)
roi = build_weight("gaussian_top_round", cfg_eff.output_size, center, cfg_eff.roi_diameter, cfg_eff.roi_softness, cfg_eff.device)
weight = threshold_weight(roi, fraction=cfg_eff.weight_threshold)
target_amp = target_amp * weight
target_amp = target_amp * torch.sqrt(beam.square().sum() / target_amp.square().sum().clamp_min(1e-9))

phase = torch.nn.Parameter(2.0 * torch.pi * torch.rand((cfg_eff.slm_size, cfg_eff.slm_size), device=device))
optimizer = torch.optim.AdamW([phase], lr=0.03)

best_eff = 0.0
for step in range(300):
    optimizer.zero_grad()
    out_amp, out_phase = propagator(beam, phase)
    overlap = normalized_overlap(target_amp, target_phase, out_amp, out_phase, weight)
    eff = efficiency_metric(weight, out_amp)
    loss = (1.0 - overlap).square() + 0.5 * (1.0 - eff).square()
    loss.backward()
    optimizer.step()
    if eff.item() > best_eff:
        best_eff = eff.item()
    if step % 50 == 0:
        print(f"step {step:3d}: overlap={overlap.item():.6f}  efficiency={eff.item():.6f}  best_eff={best_eff:.6f}")

print(f"\n=== Efficiency Ceiling ===")
print(f"max efficiency reached: {best_eff:.4f}")
print(f"suggested efficiency_floor (0.85 * ceiling): {0.85 * best_eff:.4f}")
````

````python
from pathlib import Path
import json
import math
import matplotlib.pyplot as plt
from IPython.display import Image, display

summary_path = Path("ai_holography") / "runs" / "benchmark_lin2025_hybrid" / "summary.json"
summary = json.loads(summary_path.read_text(encoding="utf-8"))
routes = summary.get("routes", {})
benchmark_dir = summary_path.parent
plots_dir = benchmark_dir / "plots"
plots_dir.mkdir(parents=True, exist_ok=True)


def route_order(routes: dict) -> list[str]:
    preferred = [
        "795_only",
        "795_plus_hybrid",
        "lin2025_plus_hybrid_balanced",
        "lin2025_plus_hybrid_quality",
        "lin2025_plus_feedback_proxy",
    ]
    return [name for name in preferred if name in routes]


def route_output_dir(name: str) -> Path | None:
    if name == "lin2025_plus_hybrid_quality":
        metrics = routes.get(name, {})
        selected = metrics.get("selected_checkpoint")
        ckpt = Path(selected).stem if selected else None
        variant = metrics.get("quality_variant")
        if ckpt and variant:
            return Path("ai_holography") / "runs" / f"benchmark_quality_flat_{ckpt}_{variant}" / "outputs" / f"flat_lin2025_plus_hybrid_quality_{ckpt}_{variant}"
    fallback = {
        "795_only": Path("ai_holography") / "runs" / "benchmark_795_flat" / "outputs" / "flat_795_only",
        "795_plus_hybrid": Path("ai_holography") / "runs" / "benchmark_795_hybrid_flat" / "outputs" / "flat_795_plus_hybrid",
        "lin2025_plus_hybrid_balanced": Path("ai_holography") / "runs" / "benchmark_balanced_flat" / "outputs" / "flat_lin2025_plus_hybrid_balanced",
        "lin2025_plus_feedback_proxy": Path("ai_holography") / "runs" / "benchmark_feedback_proxy_flat" / "outputs" / "flat_lin2025_plus_feedback_proxy_round1",
        "lin2025_plus_hybrid_quality": Path("ai_holography") / "runs" / "benchmark_quality_flat" / "outputs" / "flat_lin2025_plus_hybrid_quality",
    }
    return fallback.get(name)


order = route_order(routes)
labels = {
    "795_only": "795 only",
    "795_plus_hybrid": "795+hybrid",
    "lin2025_plus_hybrid_balanced": "balanced",
    "lin2025_plus_hybrid_quality": "quality",
    "lin2025_plus_feedback_proxy": "feedback proxy",
}
core_uniformity = [float(routes[name].get("core_uniformity_loss", float("nan"))) for name in order]
core_phase = [float(routes[name].get("core_phase_flatness", float("nan"))) for name in order]
efficiency = [float(routes[name].get("efficiency", float("nan"))) for name in order]
efficiency_floor = float(summary.get("routes", {}).get(summary.get("best_route", ""), {}).get("efficiency_floor", 0.05))

x = list(range(len(order)))
fig, axes = plt.subplots(2, 1, figsize=(12, 10), constrained_layout=True)
width = 0.35
axes[0].bar([i - width / 2 for i in x], core_uniformity, width=width, label="Core uniformity loss", color="#5B4DB2")
axes[0].bar([i + width / 2 for i in x], core_phase, width=width, label="Core phase flatness", color="#1F9D8A")
axes[0].set_xticks(x, [labels[n] for n in order])
axes[0].set_ylabel("Loss (lower=better)")
axes[0].set_title("Route Comparison: Core Metrics")
axes[0].grid(axis="y", alpha=0.25)
axes[0].legend()

axes[1].bar(x, efficiency, width=0.55, color="#3A86E9", label="Efficiency")
axes[1].axhline(efficiency_floor, color="#D84A4A", linestyle="--", linewidth=1.5, label=f"Floor = {efficiency_floor:.3f}")
axes[1].set_xticks(x, [labels[n] for n in order])
axes[1].set_ylabel("Efficiency (higher=better)")
axes[1].set_ylim(0.0, max(1.0, max(efficiency + [efficiency_floor]) * 1.15))
axes[1].set_title("Route Comparison: Efficiency (diagnostic only)")
axes[1].grid(axis="y", alpha=0.25)
axes[1].legend()

metrics_plot = plots_dir / "route_metrics_overview.png"
fig.savefig(metrics_plot, dpi=180, bbox_inches="tight")
plt.close(fig)
print("saved:", metrics_plot)
display(Image(filename=str(metrics_plot)))

# 显示最佳路由的图片
print()
print("=== Best Route Images ===")
best_route = summary["best_route"]
out_dir = route_output_dir(best_route)
if out_dir is not None:
    for name in ["summary.png", "linecuts.png"]:
        p = out_dir / name
        if p.exists():
            display(Image(filename=str(p)))

existing = []
for name in order:
    base = route_output_dir(name)
    if base is None:
        continue
    summary_img = base / "summary.png"
    linecuts_img = base / "linecuts.png"
    if summary_img.exists() or linecuts_img.exists():
        existing.append((name, summary_img if summary_img.exists() else None, linecuts_img if linecuts_img.exists() else None))

for name, summary_img, linecuts_img in existing:
    label = labels[name]
    variant = routes.get(name, {}).get("quality_variant")
    if variant:
        label = f"{label} ({variant})"
    print()
    print(f"== {label} ==")
    if summary_img is not None:
        display(Image(filename=str(summary_img)))
    if linecuts_img is not None:
        display(Image(filename=str(linecuts_img)))
````

## Final Export
Run this cell at the very end. It copies the trained checkpoints, benchmark `summary.json`, and the whole `plots/` directory into `colab_export_bundle/`, then builds a zip archive automatically.

Suggested local use after download:
1. put checkpoints under `lin2025_holography/checkpoints/wider_net_v60_phaseflat_uniform/`;
2. use png files in `benchmark_plots/` directly for PPT replacement;
3. use `benchmark_summary.json` for result tables, Pareto plots, and conclusion slides;
4. use `export_manifest.json` to verify version, source paths, and bundled files.



````python
from pathlib import Path
import json
import shutil
import numpy as np
import torch

from ai_holography.config import HolographyConfig
from ai_holography.pipeline import AIHolographyPipeline
from ai_holography.references import (
    apply_795_reference_metadata,
    find_795_reference_files,
    parse_795_reference_name,
)
from lin2025_holography.config import Lin2025Config
from lin2025_holography.predictor import predict_hologram_phase


EXPORT_DIR = Path("colab_export_bundle")
BENCHMARK_DIR = Path("ai_holography") / "runs" / "benchmark_lin2025_hybrid"
SUMMARY_PATH = BENCHMARK_DIR / "summary.json"
PLOTS_DIR = BENCHMARK_DIR / "plots"
QUALITY_CHECKPOINT_SUBDIR = globals().get("QUALITY_CHECKPOINT_SUBDIR", "wider_net_v60_phaseflat_uniform")
lin_ckpt_dir = Path("lin2025_holography") / "checkpoints" / QUALITY_CHECKPOINT_SUBDIR

if EXPORT_DIR.exists():
    shutil.rmtree(EXPORT_DIR)
EXPORT_DIR.mkdir(parents=True, exist_ok=True)

for name in [
    "lin2025_best_supervised.pt",
    "lin2025_best_quality.pt",
    "lin2025_best_hybrid_polish.pt",
    "lin2025_training_history.json",
]:
    src = lin_ckpt_dir / name
    if src.exists():
        shutil.copy2(src, EXPORT_DIR / name)

refs = find_795_reference_files(Path("ftl_gen"))
device = "cuda" if torch.cuda.is_available() else "cpu"
lin_cfg = Lin2025Config(
    target_mode="flat_top",
    slm_size=256,
    device=device,
    checkpoint_dir=lin_ckpt_dir,
)


def build_problem_from_ref(style: str):
    for ref in refs:
        info = parse_795_reference_name(ref)
        if info.get("style") == style:
            cfg = HolographyConfig(run_name=f"export_{style}")
            cfg.device = device
            cfg.reference_phase_path = ref
            cfg = apply_795_reference_metadata(cfg, info)
            cfg.phase_type = "phase_flat"
            pipe = AIHolographyPipeline(cfg)
            return ref, pipe.build_problem()
    raise FileNotFoundError(style)


manifest = {
    "notebook_version": "v82",
    "project_root": str(Path.cwd()),
    "checkpoint_dir": str(lin_ckpt_dir),
    "export_dir": str(EXPORT_DIR),
    "files": [],
    "placement_instructions": {
        "local_checkpoint_dir": "D:/Trae products/flat_top light/lin2025_holography/checkpoints/<your_autodl_run_name>/",
        "local_analysis_script": "D:/Trae products/flat_top light/colab_local_split/analyze_lin2025_checkpoint_local.py",
    },
}

prediction_targets = [
    ("round_top", "lin2025_best_supervised.pt"),
    ("flat_top", "lin2025_best_supervised.pt"),
]

for style, checkpoint_name in prediction_targets:
    ref, problem = build_problem_from_ref(style)
    checkpoint_path = lin_ckpt_dir / checkpoint_name
    if not checkpoint_path.exists():
        continue
    pred_phase, loaded = predict_hologram_phase(
        lin_cfg,
        target_amp=problem["target_amp"],
        target_phase=problem["target_phase"],
        checkpoint=checkpoint_path,
    )
    out_name = f"pred_{style}_phase.npy"
    np.save(EXPORT_DIR / out_name, pred_phase.detach().cpu().numpy().astype(np.float32))
    manifest["files"].append(
        {
            "name": out_name,
            "source_reference": str(ref),
            "checkpoint_used": str(loaded) if loaded is not None else str(checkpoint_path),
        }
    )

if SUMMARY_PATH.exists():
    shutil.copy2(SUMMARY_PATH, EXPORT_DIR / "benchmark_summary.json")
    manifest["files"].append({"name": "benchmark_summary.json", "source": str(SUMMARY_PATH)})

plots_export_dir = EXPORT_DIR / "benchmark_plots"
if PLOTS_DIR.exists():
    shutil.copytree(PLOTS_DIR, plots_export_dir)
    manifest["files"].append({"name": "benchmark_plots", "source": str(PLOTS_DIR)})

(EXPORT_DIR / "export_manifest.json").write_text(json.dumps(manifest, indent=2), encoding="utf-8")
zip_path = shutil.make_archive(str(EXPORT_DIR), "zip", root_dir=EXPORT_DIR.parent, base_dir=EXPORT_DIR.name)
print("exported bundle dir:", EXPORT_DIR)
print("exported zip:", zip_path)
print("bundle contents:")
for p in sorted(EXPORT_DIR.rglob("*")):
    rel = p.relative_to(EXPORT_DIR)
    print("-", rel)

````
